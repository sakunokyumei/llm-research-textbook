## 過去問と本番：データを分ける

モデルの点数をつける前に、大事な準備があります。データを2つに分けるのです。

> **💡 理論のポイント：なぜ分けるの？**
>
> テスト勉強で、過去問の**答えを丸暗記**した人を想像してください。同じ過去問なら満点ですが、本番で少し違う問題が出たら解けません。
>
> モデルも同じです。数えるのに使った文章で点数をつけると、「丸暗記」がうまいほど高得点になってしまいます。本当に知りたいのは「**見たことのない文章**でも、次の文字を当てられるか」です。
>
> そこで、データを分けます。
>
> - **訓練データ**（training data）：モデルが学ぶ（数える）ためのデータ。過去問
> - **検証データ**（validation data）：点数をつけるためだけのデータ。モデルには絶対に見せない。本番の模試
>
> さらに研究では、最後の最後に一度だけ使う **テストデータ** をもう1つ取っておくこともあります（第20章）。

訓練データだけでbigramを数え直します。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
pairs = Counter(zip(train_text, train_text[1:]))
firsts = Counter(train_text[:-1])

def prob(a, b):
    """P(b | a)：「a」の次が「b」である確率"""
    if firsts[a] == 0:
        return 0.0
    return pairs[(a, b)] / firsts[a]

print("P(生|先) =", prob("先", "生"))
print("P(は|生) =", prob("生", "は"))
````

## 文章の確率を計算する

モデルが「この文章はありそうだ」と思うほど、良いモデルのはずです。では、文章全体の確率（同時確率）はどう計算するのでしょう？ 数学的に**厳密な等式**と、bigramモデルの**近似**をしっかり区別して押さえましょう。

#### 📐 数学のポイント：厳密な「確率の連鎖律」 vs bigramの「マルコフ近似」

任意の文字列 $(x_1, x_2, x_3)$ が続けて出てくる確率は、条件付き確率の定義から**常に厳密に**次の掛け算に分解できます（**確率の連鎖律**、chain rule of probability）。

$$P(x_1, x_2, x_3) = P(x_1) \cdot P(x_2 \mid x_1) \cdot P(x_3 \mid x_1, x_2)$$

「先生は」なら、厳密には $P(\text{先生は}) = P(\text{先}) \cdot P(\text{生} \mid \text{先}) \cdot P(\text{は} \mid \text{先生})$ です。本物のLLM（GPTなど）は、3文字目を予測するときに**それまでの全文脈 $(x_1, x_2)$ を条件にする**この厳密な連鎖律をそのまま使います。

一方、bigramモデルは「直前の1文字だけを見る」という**マルコフ近似**（$P(x_3 \mid x_1, x_2) \approx P(x_3 \mid x_2)$）を置きます。

$$P_{\text{bigram}}(x_1, x_2, x_3) = P(x_1) \cdot P(x_2 \mid x_1) \cdot P(x_3 \mid x_2)$$

最初の1文字 $P(\text{先})$ を除いた遷移確率の積で「先生は」を計算すると、bigramの近似値は次のようになります。

$$P_{\text{bigram}}(\text{生}, \text{は} \mid \text{先}) = P(\text{生} \mid \text{先}) \times P(\text{は} \mid \text{生}) \approx 0.886 \times 0.231 \approx 0.205$$

**区別が大切**：上の $P(x_1, x_2, x_3) = P(x_1)P(x_2 \mid x_1)P(x_3 \mid x_1, x_2)$ は**常に成り立つ数学の恒等式**、$P(x_3 \mid x_1, x_2) \approx P(x_3 \mid x_2)$ は**計算を簡単にするためのモデル上の近似**です。

では、訓練データの最初の数百文字で bigram の遷移確率を掛け続けてみましょう。どのペアも訓練データで数えたものなので、確率が0になることはないはずです。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
p = 1.0
for i, (a, b) in enumerate(zip(train_text, train_text[1:])):
    p = p * prob(a, b)
    if i in [10, 50, 100, 150, 200, 300]:
        print(f"{i:4d}文字目まで：確率 {p}")
````

## 救世主：対数（log）

ここで登場するのが **対数**（たいすう、logarithm）です。記号を丸暗記するのではなく、「なぜ対数が必要なのか」から9ステップでたどりましょう。

#### 📐 数学のポイント：対数 $\log$：9ステップで理由から理解する

**① 解決したい具体的な問題**：何千文字もの確率の掛け算 $p_1 \times p_2 \times \dots \times p_N$ を、0に潰れず（アンダーフローせず）、かつ**モデル間の大小関係を変えずに**計算したい。

**② 素朴な掛け算ではなぜダメなのか？**：確率は必ず $0 \le p \le 1$ なので、掛けるたびに桁が急減します。300文字も掛けると $10^{-308}$ を下回り、64ビット浮動小数点数（`float`）では `0.0` に丸められて全モデルの点数が同じ `0.0` になってしまいます。

**③ 新しい発想**：「数そのもの」ではなく、**「ある基準の数を何回掛けたか（指数の肩の数）」**で数える世界へ引っ越す。指数の世界では**掛け算が足し算に化ける**（$10^2 \times 10^3 = 10^{2+3}$）ので、何万回足しても0に潰れません！

**④ 記号の読み方表**：
| 記号 | 読み方 | 意味 |
|---|---|---|
| $\log_{10} x$ | 「ログ10の $x$」 | $10$ を何回掛けたら $x$ になるか（常用対数） |
| $\log x$（または $\ln x$） | 「ログ $x$」（自然対数） | 第3章の $e \approx 2.718$ を何回掛けたら $x$ になるか。Pythonは `math.log(x)` |
| $e^x$ と $\log x$ | 互いに逆の操作 | $\log(e^x) = x$、$e^{\log y} = y$（$y > 0$） |

**⑤ 小さな数字で手計算**：$\log_{10} 1000$ は「10を何回掛けたら1000になるか」という意味で、答えは `3` です（$10 \times 10 \times 10 = 1000$）。

| $x$ | 0.001 | 0.01 | 0.1 | 1 | 10 | 100 | 1000 |
|---|---|---|---|---|---|---|---|
| $\log_{10} x$ | -3 | -2 | -1 | 0 | 1 | 2 | 3 |

1より小さい数はマイナスになります（`0.01` は「10で2回割った」＝ $10^{-2}$ なので `-2`）。掛け算も検算してみましょう：$0.1 \times 0.01 = 0.001$ ですが、対数にすると $(-1) + (-2) = -3 = \log_{10}(0.001)$ とぴったり一致します！

**⑥ 一般式（自然対数）**：機械学習では10の代わりに $e \approx 2.718$ を底にした **自然対数** $\log$ を使います（第3章の `softmax` に出てくる $e^x$ のちょうど逆関数だからです）。
$$\log(p_1 \times p_2 \times \dots \times p_N) = \log p_1 + \log p_2 + \dots + \log p_N = \sum_{i=1}^{N} \log p_i$$

**⑦ Pythonでの検算**（下のセルで実際に確かめます）：`math.log(math.e) == 1.0`、`math.log(1) == 0.0`、`math.log(0.5) ≈ -0.693`。

## 損失：驚きの平均

「log確率の合計」は文章が長いほどマイナスに大きくなるので、長さの違う文章どうしを比べにくいです。そこで、**1文字あたりの平均**にします。さらに、マイナスの数は扱いにくいので、**符号をひっくり返して**プラスにします。これが言語モデルの点数、**損失**（loss）です。

$$L = -\frac{1}{N}\sum_{i=1}^{N} \log p_i = -\frac{1}{N}\sum_{i=1}^{N} \log P(\text{次の文字}_i \mid \text{今の文字}_i)$$

#### 📐 数学のポイント：なぜ $1-p$ ではなく $-\log p$ なのか？：9ステップで理由から理解する

**① 解決したい具体的な問題**：正解文字に対する予測確率 $p_1, p_2, \dots, p_N$ から、「外したほど大きい罰（損失）」を1つの数 $L$ にまとめたい。

**② 素朴な方法（$1 - p$ や $(1 - p)^2$）ではなぜダメなのか？**
「外れ度合いなら $1 - p$ や $(1 - p)^2$ でよさそう」と思えますよね。実際に極端な外れ値で比べてみましょう。

| モデルが正解につけた確率 $p$ | $1 - p$ | $(1 - p)^2$ | $-\log p$ |
|---|---|---|---|
| $1.0$（完璧に的中） | $0.00$ | $0.00$ | $0.00$ |
| $0.5$（五分五分） | $0.50$ | $0.25$ | $0.69$ |
| $0.01$（1%と軽視） | $0.99$ | $0.9801$ | $4.61$ |
| $0.0001$（0.01%と決めつけて大外れ） | $0.9999$ | $0.9998$ | $9.21$ |
| $0$（絶対に起きないと断言して大外れ） | $1.00$ | $1.00$ | $\infty$（無限大！） |

- **欠点1（大外れに甘すぎる）**：$1 - p$ や $(1 - p)^2$ では「確率1%」と「確率0.01%」の罰が `0.99` vs `0.9999` でほとんど差がつかず、「絶対に起きない（$p=0$）」と断言して外してもたかだか `1.0` の減点で済んでしまいます。一方 $-\log p$ は $p$ が0に近づくほど急激に重い罰を与え、自信満々な大外れを徹底的に矯正します。
- **欠点2（同時確率の積と一致しない）**：ここで**総乗記号 $\prod$（パイ）**を押さえておきましょう。第3章の $\sum$（全部足す）の掛け算版で、$\prod_{i=1}^{N} p_i = p_1 \times p_2 \times \dots \times p_N$（Pythonでは `prod = 1.0` から `for p in probs: prod *= p`）を意味します。$1 - p$ の平均を小さくしても、文章全体の同時確率 $\prod_{i=1}^N p_i$ を最大にすることとは一致しません。**$-\log p$ の平均だけを小さくすることが、同時確率 $\prod_{i=1}^N p_i$ の最大化（最尤推定）と数学的にぴったり一致**します（$-\frac{1}{N}\log \prod_{i=1}^N p_i = -\frac{1}{N}\sum_{i=1}^N \log p_i$）。
- **先取りの利点（第8章）**：第3章の `softmax` と $-\log p$ を組み合わせると、学習に使う微分（勾配）が $p - y$ という最もシンプルで壊れにくい形になります！

**③ 記号の読み方表**：
| 記号 | 読み方・役割 |
|---|---|
| $L$ | 損失（Loss）。小さいほど良い（0が完璧） |
| $-$ | 負の対数 $\log p_i \le 0$ の符号を反転し、0以上の「驚き」にする |
| $\frac{1}{N}$ | ペア総数 $N$ で割り、文章の長さに依らない「1文字あたりの平均」にする |
| $\sum_{i=1}^{N}$ | $i=1$ から $N$ まで全ペアの対数確率を足し合わせる（第3章の $\sum$） |
| $p_i$ | $i$ 番目の正解文字にモデルが割り当てた予測確率 $P(\text{次}_i \mid \text{今}_i)$ |

**④ 小さな数字で手計算**：3ステップの正解確率が $p = [0.5,\; 0.5,\; 1.0]$ のとき（$\log 0.5 \approx -0.693,\; \log 1.0 = 0$）：
$$L = -\frac{1}{3}\bigl(\log 0.5 + \log 0.5 + \log 1.0\bigr) \approx -\frac{1}{3}(-0.693 - 0.693 + 0) = \frac{1.386}{3} = 0.462$$

**⑤ Pythonの最小コードで検算**：
```python no-run

print(round(loss, 3))   # 0.462
```

#### 💡 理論のポイント：損失は「驚き」の平均

$-\log p$ は、確率 $p$ のことが起きたときの **驚きの大きさ** と考えられます。

天気予報に例えてみましょう。
- 「明日は100%晴れ」と予報して本当に晴れたら、驚きはゼロです（$-\log 1 = 0$）
- 「降水確率50%」と言われて雨が降ったら、コイントスが裏だったくらいで少し驚きます（$-\log 0.5 \approx 0.69$）
- 「降水確率1%」と言われて雨が降ったら、「えっ、降ったの！？」とかなり驚きます（$-\log 0.01 \approx 4.6$）
- 「降水確率0%（絶対に雨は降らない）」と断言していたのに土砂降りになったら、「嘘でしょ！？」と無限大にショックを受けます（$-\log 0 = \infty$）

損失は、モデルが検証データを1文字ずつ読みながら感じた「驚き」の平均です。良いモデルほど、次の文字を正しく予想できているので、驚きが少ない。だから **損失は小さいほど良い** のです。ゴルフのスコアと同じですね。

ちなみに、この量は **負の対数尤度**（negative log-likelihood、NLL）や、**交差エントロピー**（cross-entropy）とも呼ばれます。より一般的には、真の確率分布 $P$ とモデルの予測分布 $Q$ のあいだの交差エントロピーは

$$H(P, Q) = -\sum_{x} P(x)\log Q(x)$$

と定義されます。言語モデルで「正解文字 $y$ だけ確率 $P(y)=1$、他の文字はすべて $0$」（one-hot分布）のときは、和の中で正解文字の項だけが残り $H(P, Q) = -\log Q(y)$ となって負の対数尤度とぴったり一致します。また、真の分布自身の不確かさである **エントロピー** $H(P) = -\sum_{x} P(x)\log P(x)$ と、2つの分布のズレを測る **KLダイバージェンス** $D_{\mathrm{KL}}(P \| Q) = \sum_x P(x)\log\frac{P(x)}{Q(x)}$（第22章・[付録D](https://sakunokyumei.github.io/llm-research-textbook/d-math.html)）を使うと、$H(P, Q) = H(P) + D_{\mathrm{KL}}(P \| Q)$ という関係が成り立ちます。$H(P)$ はデータ固有の定数なので、交差エントロピーを最小化することは、モデルの予測分布 $Q$ を真の分布 $P$ に近づけることと同じ意味になります。

- **ヒント 2**：4つの驚き `0.693`, `1.386`, `0.693`, `0` をすべて足し合わせ、ステップ数 `4` で割りましょう。

## 「絶対」を言わないモデル：スムージング

対策はシンプルです。**どのペアにも、ほんの少しだけ回数を足しておく**のです。見たことがないペアにも「0.01回くらいは見たかもね」とおまけをつける。これを **スムージング**（smoothing、なめらかにする）と言います。

$$P(b \mid a) = \frac{\text{「ab」の回数} + k}{\text{「a」の回数} + k \times V}$$

$V$ は語彙の数です。分母にも $k \times V$ を足すのは、確率の合計をちゃんと1にするためです（全部の $b$ に $k$ ずつ足したので）。

#### 📐 数学のポイント：小さな手計算で確かめる（$V = 4$、$k = 1$）

語彙が `あ, い, う, え`（$V = 4$）だけの世界で、「あ」の出現が6回（内訳：`あい:3, あう:2, ああ:1, あえ:0`）だったとします。全4文字に $k = 1$ ずつ足すと、分母も $k \times V = 1 \times 4 = 4$ 増えて $6 + 4 = 10$ になります。

| 次の文字 $b$ | 元の回数 | $+k$ 後の分子 | スムージング後の確率 $P(b \mid \text{あ})$ |
|---|---|---|---|
| い | 3 | $3+1=4$ | $4/10 = 0.4$ |
| う | 2 | $2+1=3$ | $3/10 = 0.3$ |
| あ | 1 | $1+1=2$ | $2/10 = 0.2$ |
| え（未出現！） | 0 | $0+1=1$ | $1/10 = 0.1$（0にならない！） |
| **合計** | **6** | **$6 + 4 = 10$** | **$0.4 + 0.3 + 0.2 + 0.1 = 1.0$** |

もし分母に $+k \times V$ を忘れると合計が $10/6 = 1.67$ になって確率分布が壊れてしまいます。分母の $k \times V$ は「全 $V$ 文字に $k$ ずつ足した増分を帳尻合わせする項」なのです。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
V = len(set(text))

def make_smoothed(k):
    def p(a, b):
        return (pairs[(a, b)] + k) / (firsts[a] + k * V)
    return p

for k in [1, 0.1, 0.01, 0.001]:
    p_k = make_smoothed(k)
    print(f"k={k:<6} 訓練 {bigram_loss(train_text, p_k):.3f}   検証 {bigram_loss(val_text, p_k):.3f}")
````

## パープレキシティ：何択で迷っているか

損失「3.581」と言われても、ピンとこないかもしれません。そこで、もう少し直感的な数に変換します。

$$\text{パープレキシティ} = e^{\text{損失}}$$

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

#### 📐 数学のポイント：パープレキシティ $\mathrm{PPL} = e^L$ の導出・手計算・実装・落とし穴

**① なぜ $e^L$ が「平均何択」になるのか？（導出）**
各位置 $i$ で正解に割り当てた確率が $p_i$ なら、その位置の「逆数 $1/p_i$」は「その確率 $p_i$ になる均等な選択肢の数」（例：$p_i = 0.25$ なら $1/0.25 = 4$ 択）を表します。全 $N$ 文字にわたる「選択肢の数の幾何平均」を計算して対数・指数で書き換えると、ちょうど $e^L$ に一致します！

$$\mathrm{PPL} = \left(\prod_{i=1}^{N} \frac{1}{p_i}\right)^{1/N} = \exp\!\left(\frac{1}{N}\sum_{i=1}^{N} \log\frac{1}{p_i}\right) = \exp\!\left(-\frac{1}{N}\sum_{i=1}^{N}\log p_i\right) = e^L$$

**② 小さな手計算（均等な例 ＆ ばらつきがある例）**
- 毎回均等に4択（$p = 1/4 = 0.25$）：$L = -\log(1/4) = \log 4 \approx 1.386 \;\Rightarrow\; \mathrm{PPL} = e^{\log 4} = 4$（ちょうど4択）
- 毎回100%的中（$p = 1.0$）：$L = -\log 1 = 0 \;\Rightarrow\; \mathrm{PPL} = e^0 = 1$（迷いゼロの1択）
- **確率がばらつく例**：さきほど手計算した $p = [0.5,\; 0.5,\; 1.0]$ のとき
  - 幾何平均から：$(2 \times 2 \times 1)^{1/3} = 4^{1/3} \approx 1.587$
  - 損失 $L \approx 0.462$ から：$e^{0.462} \approx 1.587$（完全に一致！）

**③ 再利用できる最小のPython実装（ゼロ確率・オーバーフロー対策つき）**
```python no-run
import math

def perplexity(probs):
    if any(p <= 0 for p in probs):
        return float("inf")             # 確率0が1つでもあれば PPL = ∞
    loss = -sum(math.log(p) for p in probs) / len(probs)
    if loss > 700:                      # float64 は exp(709.78...) 超で OverflowError
        return float("inf")
    return math.exp(loss)

print(round(perplexity([0.5, 0.5, 1.0]), 3))   # 1.587
print(perplexity([0.25, 0.25, 0.25, 0.25]))    # 4.0
print(perplexity([0.5, 0.0]))                  # inf
```

> **💡 理論のポイント：パープレキシティの直感と「3つの重要な注意点」**
>
> 2061択から約36択へ。数えるだけで、迷いを約57分の1にできたわけです。ただし、パープレキシティを研究や評価で使うときは、次の3つの落とし穴に必ず注意してください。
>
> 1. **「実質 $K$ 択」のたとえの限界**：$\mathrm{PPL} = 10$ は「どの文字でも常に10個が均等に並んでいる」という意味ではなく、あくまで**幾何平均としての等価な選択肢数**です。たとえば99文字を $p \approx 1.0$ で当てても、1文字だけ $p = 10^{-100}$ と大外れすれば全体の $\mathrm{PPL}$ は一気に $10$ へ跳ね上がります（$p=0$ が1つ混ざるだけで $\mathrm{PPL} = \infty$）。
> 2. **トークナイザー（語彙・分割単位）が違うモデルどうしは直接比べられない**：1文字単位とサブワード（BPE、第14章）単位では1トークンあたりの情報量も $N$ も異なるため、生の $\mathrm{PPL}$ の大小を比べることはできません（第20章で詳しく検証します）。
> 3. **評価データや前処理が違う数字は比べられない**：同じモデルでも、易しい文か難しい文か、ルビ除去などの前処理が違うかで $\mathrm{PPL}$ は大きく変わります。公平な比較には**まったく同じ評価データ・前処理・トークナイザー**が必要です。

**考え方**：「損失の数字」をパープレキシティで直感化し、同時に検証データでのトレードオフ（知識の薄まり vs 未出現への驚き）を言葉で説明できれば、第5章の核心は完全に定着しています。

1. パープレキシティ比較：2061択 → 35.9択（迷いが約1/57に減少）。
1. `k` が大きすぎる（`k=1`）：おまけが多くて数えた知識が薄まり悪化（4.385）。
1. `k` が小さすぎる（`k=0.001`）：未出現ペアの確率が極小になり $-\log p$ が大きく増えて悪化（3.652）。

<details><summary><b>もっと深く：情報理論との関係（ビットで数える）</b>（クリックで開く）</summary>

自然対数の代わりに $\log_2$ を使うと、損失の単位は **ビット** になります。「1文字を予測するのに、平均何ビット分の情報が足りないか」という意味で、**エントロピー**という考え方につながります。$\log_2 x = \log x / \log 2$ なので、3.581 ÷ 0.693 ≈ 5.17ビット。つまり私たちのモデルは、次の1文字を当てるのに「Yes/Noの質問をあと約5回したい」状態です。<br>
この考え方は、1948年にクロード・シャノンが作った **情報理論** から来ています。シャノンは英語の1文字あたりのエントロピーを、人間に次の文字を当てさせる実験で推定しました。まさに人間を言語モデルとして使ったわけです。
