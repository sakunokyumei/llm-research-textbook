````python
import re, math, time, urllib.request
from collections import Counter
import torch

def load_soseki(file_id="773_ruby_5968"):
    url = ("https://raw.githubusercontent.com/aozorahack/aozorabunko_text/0984f7dc18a2270bdfd2ff99bf14fd1b3ebb4f65/"
           f"cards/000148/files/{file_id}/{file_id}.txt")
    try:
        with urllib.request.urlopen(url, timeout=30) as res:
            text = res.read().decode("cp932")
    except Exception as e:
        raise RuntimeError(f"青空文庫の取得に失敗しました（時間をおいて再実行するか付録Eを参照）: {e}") from e
    text = text.split("-------------------------------------------------------")[2]
    text = text.split("底本：")[0]
    text = re.sub(r"《.*?》", "", text)
    text = re.sub(r"※?［＃.*?］", "", text)
    text = text.replace("｜", "").replace("\r\n", "\n")
    text = re.sub(r"\n{3,}", "\n\n", text)
    return text.strip()

text = load_soseki()
chars = sorted(set(text))
V = len(chars)
stoi = {ch: i for i, ch in enumerate(chars)}
itos = {i: ch for ch, i in stoi.items()}
print("語彙の数：", V)
````

#### 💡 理論のポイント：ベクトル・行列・テンソルと「形」

- 数を1列に並べたもの：**ベクトル**（vector）。`a` は長さ3のベクトル
- 数を表（縦×横）に並べたもの：**行列**（matrix）。`M` は2行3列の行列
- さらに何枚も重ねたもの（3次元以上も含めた総称）：**テンソル**（tensor）

| 次元数（ランク） | 呼び名 | 具体例 | `.shape` | 要素数 `.numel()` |
|---|---|---|---|---|
| 0次元 | スカラー（数1個） | `torch.tensor(6)` | `[]` | $1$ |
| 1次元 | ベクトル | `torch.tensor([1, 2, 3])` | `[3]` | $3$ |
| 2次元 | 行列（表） | `M`（2行3列） | `[2, 3]` | $2 \times 3 = 6$ |
| 3次元 | 3次元テンソル | 第3章の `out.logits` | `[1, 6, 151936]` | $1 \times 6 \times 151936$ |

**多次元インデックスの読み方**（すべて **0始まり**、`[行, 列]` の順）：
- `M[0]` → 0行目ぜんぶ：`[1, 2, 3]`（形 `[3]`）
- `M[:, 1]` → 全行の1列目：`[2, 5]`（形 `[2]`。`:` は「その軸を全部」の意味）
- `M[1, 2]` → 1行目の2列目：`6`（形 `[]`）

テンソルを扱うときに一番大事なのは **形**（shape）です。この先、エラーの半分以上は「形が合わない」ことが原因です。迷ったら、とにかく `.shape` を `print` しましょう。これはこの教科書で一番役に立つアドバイスかもしれません。

> **📘 基本情報とのつながり：2次元配列**
>
> 基本情報の擬似言語やアルゴリズムの問題で出てくる「2次元配列」が、まさに行列です。`M[1, 2]` は「1行目の2列目」。ただしPythonは0から数えるので、人間の感覚では「2行目の3列目」です。試験の問題が1から数えるか0から数えるかは、毎回確認しましょう。

## bigramの表を作る

いよいよ本題です。第4章の「今の文字（行）×次の文字（列）」の表を、2061×2061の行列として作ります。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
n = int(len(text) * 0.9)
train_text, val_text = text[:n], text[n:]

pairs = Counter(zip(train_text, train_text[1:]))
N = torch.zeros((V, V))
for (a, b), count in pairs.items():
    N[stoi[a], stoi[b]] = count

print(N.shape)
print("「先」→「生」の回数：", N[stoi["先"], stoi["生"]].item())
print("表のマスの数：", N.numel(), " うち0でないマス：", (N > 0).sum().item())
````

> **🐍 Python・PyTorchのポイント：.item() と比較**
>
> `N[i, j]` の結果は「数が1つだけ入ったテンソル」です。`.item()` を付けると、ふつうのPythonの数として取り出せます。<br>
> `N > 0` は、全部のマスについて「0より大きい？」を調べた、True/Falseの表を作ります。それを `.sum()` すると、Trueの数（＝0でないマスの数）が数えられます。

約420万マスのうち、0でないのは約2万マス（0.5%以下）。第4章で話した「スパース（まばら）」が、数字で見えました。

| 属性 / メソッド | 具体例と出力 | 型 | 役割 |

```python no-run
import torch
t_int = torch.tensor([1, 2, 3])          # dtype: torch.int64（ID・インデックス用）
```

また、テンソルの総要素数を保ったまま形を変える操作に **`.view(...)`** と **`.reshape(...)`** があります（第9章以降で多用します）。

```python no-run
x = torch.arange(6).reshape(2, 3)        # 形 [6] → [2, 3]
xt = x.transpose(0, 1)                   # 軸を入れ替えて [3, 2]（メモリ上は不連続になる）
# xt.view(6)                             # RuntimeError! メモリが不連続だと .view() は失敗する
print(xt.reshape(6))                     # OK! 必要なら自動でコピーして [6] に変形
print(xt.contiguous().view(6))           # OK! 連続化してから .view() しても同じ
```

- **`.view(...)`**：メモリをコピーせず高速に形だけ変えるが、メモリが連続（contiguous）でないとエラーになる。
- **`.reshape(...)`**：連続なら `.view()` と同じ（コピーなし）、不連続なら自動でコピーして安全に変形してくれる。迷ったら `.reshape()` を使うと事故を防げます（[付録B](https://sakunokyumei.github.io/llm-research-textbook/b-python.html)にも整理あり）。

#### 📝 確認問題（`ch06-q04` · Lv.2 · .shape・.dtype・.item() と条件マスク (N > 0).sum()）

本文のコード `(N > 0).sum().item()` がやっている処理の正しい順番と、最終的な戻り値の型はどれですか？

> **💡 理論のポイント：合計の方向と、ブロードキャスト**
>
> `sum(dim=1)` は「1番目の軸（横方向）に沿って足す」、つまり**各行の合計**です。`dim=0` なら各列の合計になります。`keepdim=True` は「足した軸を消さずに長さ1で残す」指定で、結果の形は `[2061, 1]`（縦に2061個並んだ数）になります。
>
> - `dim=0`：縦方向に足す（0番目の軸をつぶす → 各**列**の合計になる）
> - `dim=1`：横方向に足す（1番目の軸をつぶす → 各**行**の合計になる）
>
> 2行3列の具体例 $M = \begin{pmatrix} 1 & 2 & 3 \\ 4 & 5 & 6 \end{pmatrix}$（形 `[2, 3]`）で手計算してみましょう：
> - `M.sum(dim=0)` → 縦に足す：`[1+4, 2+5, 3+6] = [5, 7, 9]`（形 `[3]`）
> - `M.sum(dim=1)` → 横に足す：`[1+2+3, 4+5+6] = [6, 15]`（形 `[2]`）
> - `M.sum(dim=1, keepdim=True)` → 横に足し、軸1を長さ1で残す：`[[6], [15]]`（形 `[2, 1]`）
>
> そして `(N + k) / row_sums` では、形が `[2061, 2061]` と `[2061, 1]` で違うのに、割り算ができています。これは **ブロードキャスト** という仕組みで、長さ1の軸が自動的に「横にコピーされた」ものとして扱われるからです。
>
> <div class="math-display">[[a, b, c],      [[x],        [[a/x, b/x, c/x],
>  [d, e, f]]  ÷   [y]]    →    [d/y, e/y, f/y]]</div>
>
> 各行が、その行の合計で割られます。for文なしで、420万マスの割り算が一瞬です。

<details><summary><b>基礎の補足：ステップで理解するブロードキャストの「右揃えルール」</b>（クリックで開く）</summary>

2つのテンソルの形が違うとき、PyTorchは次の2ステップで計算できるか判定します。
1. **2つの形を右端にそろえて並べる**（足りない左側には `1` を補って考える）
2. **各軸について「数字が一致する」または「どちらかが `1`」ならOK**（`1` の側がもう片方の長さにコピーされる）

| 左の形 | 右の形 | 右揃えにした姿 | 判定と結果の形 |
|---|---|---|---|
| `[2, 3]` | `[2, 1]` | `(2, 3)` と `(2, 1)` | **OK** → `[2, 3]`（右の軸1が横に3倍コピー） |
| `[2, 3]` | `[3]` | `(2, 3)` と `(1, 3)` | **OK** → `[2, 3]`（右の軸0が縦に2倍コピー） |
| `[2, 3]` | `[2]` | `(2, 3)` と `(1, 2)` | **ERROR**（右端が `3` と `2` で不一致、かつどちらも `1` でない） |
| `[2, 1]` | `[1, 3]` | `(2, 1)` と `(1, 3)` | **OK** → `[2, 3]`（お互いに引き伸ばされる） |

</details>

#### ⚠️ つまずき注意：keepdim を忘れると「静かなバグ」が起きる！（2×2の具体例で実証）

`keepdim=True` を付けないと、`row_sums` の形は `[2061]`（1次元）になります。右揃えルールのため、正方形 `[2061, 2061]` とは次のように見なされます。

<div class="math-display">[2061, 2061] と [2061] → [2061, 2061] と [1, 2061] として扱われる</div>

小さな 2×2 行列 $N = \begin{pmatrix} 1 & 3 \\ 6 & 2 \end{pmatrix}$（各行の合計は 4 と 8）で、**正しい計算**と **`keepdim` 忘れのバグ** を手計算で比べてみましょう：

- **正しい計算（`keepdim=True`、形 `[2, 1]`＝`[[4], [8]]`）**：横にコピーされて `[[4, 4], [8, 8]]` で割る
  $$P = \begin{pmatrix} 1/4 & 3/4 \\ 6/8 & 2/8 \end{pmatrix} = \begin{pmatrix} 0.25 & 0.75 \\ 0.75 & 0.25 \end{pmatrix} \quad \text{（各行の合計：1.0, 1.0）}$$
- **バグ（`keepdim` 忘れ、形 `[2]` → `[1, 2]`＝`[[4, 8]]`）**：縦にコピーされて `[[4, 8], [4, 8]]` で**列ごと**に割ってしまう！
  $$\text{bad\_P} = \begin{pmatrix} 1/4 & 3/8 \\ 6/4 & 2/8 \end{pmatrix} = \begin{pmatrix} 0.25 & 0.375 \\ \mathbf{1.5} & 0.25 \end{pmatrix} \quad \text{（確率が1.5！各行の合計も 0.625, 1.75！）}$$

## one-hotと行列の掛け算

最後に、次の章から何度も出てくる大事な計算を1つ。**行列の掛け算**です。

文字「私」を、「私」の位置だけ1で、残りが全部0の長さ2061のベクトルで表すことにします。これを **one-hot**（ワンホット、1か所だけ熱い）ベクトルと言います。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
import torch.nn.functional as F

x = F.one_hot(torch.tensor(stoi["私"]), num_classes=V).float()
print(x.shape, x.sum())

row = x @ P            # one-hot × 確率表
print(row.shape)
print(torch.allclose(row, P[stoi["私"]]))
````

`@` は行列の掛け算の記号です。one-hotベクトルに表を掛けると、**表の「私」の行がそのまま取り出される**ことが確かめられました（`allclose` は「ほぼ等しい？」を調べる関数）。

#### 📐 数学のポイント：行列の掛け算 `@`：9ステップで理由から理解する

**① 解決したい問題**：「入力の重み付き和」や「表からの行の取り出し」を、1件ずつではなく何千件分も一気に・微分（第7章）可能な数式で計算したい。

**② 要素ごとの積 `*` と行列積 `@` の決定的な違い**：
- `A * B`：同じ位置のマスどうしを掛けるだけ（形が一致するかブロードキャストが必要）。
- `A @ B`：左の**行**と右の**列**の同じ位置どうしを掛けて足す（**内積**を全ペアで計算する）。

**③ 小さな例で手計算（`1×2 @ 2×3` と `2×2 @ 2×2`）**：
まず `1×2` と `2×3` の積です。

$$\begin{pmatrix} 1 & 2 \end{pmatrix} \begin{pmatrix} 3 & 4 & 5 \\ 6 & 7 & 8 \end{pmatrix} = \begin{pmatrix} 1\cdot3+2\cdot6 & 1\cdot4+2\cdot7 & 1\cdot5+2\cdot8 \end{pmatrix} = \begin{pmatrix} 15 & 18 & 21 \end{pmatrix}$$

2行2列どうしでも同じ手順です（左の `i` 行目 × 右の `j` 列目 ＝ 結果の `(i, j)` マス）：
$$\begin{pmatrix} 1 & 2 \\ 3 & 4 \end{pmatrix} \begin{pmatrix} 5 & 6 \\ 7 & 8 \end{pmatrix} = \begin{pmatrix} 1\cdot5+2\cdot7 & 1\cdot6+2\cdot8 \\ 3\cdot5+4\cdot7 & 3\cdot6+4\cdot8 \end{pmatrix} = \begin{pmatrix} 19 & 22 \\ 43 & 50 \end{pmatrix}$$

左の**行**と右の**列**の、同じ位置どうしを掛けて足す。これをすべての組み合わせでやります。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
A = torch.tensor([[1., 2.]])
B = torch.tensor([[3., 4., 5.],
                  [6., 7., 8.]])
print(A @ B)
print(A.shape, "@", B.shape, "→", (A @ B).shape)
````

```text
tensor([[15., 18., 21.]])
torch.Size([1, 2]) @ torch.Size([2, 3]) → torch.Size([1, 3])
```

</details>

**④ 一般式と形のルール**：
$$(AB)_{i, j} = \sum_{t=0}^{k-1} A_{i, t}\, B_{t, j}$$
- **形のルール**：$(n \times k)$ の行列と $(k \times m)$ の行列を掛けると、$(n \times m)$ になります。**真ん中の $k$ が一致していないと掛けられません**（例：`[2, 3] @ [2, 3]` はエラー、`[32, 2061] @ [2061, 64]` は `[32, 64]`）。これが「形が合わない」エラーの正体です。

**⑤ なぜ `one-hot × 行列 ＝ 行の取り出し` になるのか？（具体計算）**：
$$\begin{pmatrix} 0 & 1 & 0 \end{pmatrix} \begin{pmatrix} 10 & 20 \\ 30 & 40 \\ 50 & 60 \end{pmatrix} = 0\cdot\begin{pmatrix} 10 & 20 \end{pmatrix} + 1\cdot\begin{pmatrix} 30 & 40 \end{pmatrix} + 0\cdot\begin{pmatrix} 50 & 60 \end{pmatrix} = \begin{pmatrix} 30 & 40 \end{pmatrix}$$
one-hotベクトルを掛けると、「1の位置の行 × 1」以外は全部「× 0」で消えるので、1の位置の行がそのまま出てくるのです。

**⑥ 純粋なPythonの3重ループとの対応**：
```python
# (n x k) @ (k x m) -> (n x m) の正体は、実はこの3重ループ！
C = [[sum(A[i][t] * B[t][j] for t in range(k)) for j in range(m)] for i in range(n)]
```

> **💡 理論のポイント：なぜこれが大事なのか**
>
> 「one-hot × 表 ＝ 表の行を取り出す」。当たり前に見えますが、ここに大事な発想の転換があります。
>
> - 今は、表の中身（確率）を**数えて**決めました
> - でも、表の中身を**学習で少しずつ調整する数（パラメータ）**にしたら？
>
> 第8章では、まさにそれをやります。表の中身を最初はでたらめにして、損失が小さくなるように少しずつ調整していくのです。そのためには「調整の方向」を知る必要があります。それが次の章の **微分** です。

## GPUは何が速いのか

#### 💡 理論のポイント：並列計算

行列の掛け算は、「掛けて足す」をたくさん、**お互いに関係なく**計算できます。2061×2061の行列どうしの掛け算なら、約87億回の掛け算です。

- **CPU**：とても賢い職人が数人〜数十人。複雑な仕事が得意
- **GPU**：そこそこの作業員が数千人。単純な計算を**同時に**こなすのが得意

行列の掛け算は「単純な計算を大量に」の典型なので、GPUの独壇場です。LLMの計算のほとんどは行列の掛け算なので、LLMにはGPUが欠かせないのです。Colabで確かめたい人は、次を実行してみてください（CPUのままなら `cuda` の行は飛ばされます）。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
X = torch.randn(2000, 2000)
start = time.time(); _ = X @ X; print(f"CPU：{time.time() - start:.3f}秒")
if torch.cuda.is_available():
    Xg = X.to("cuda"); _ = Xg @ Xg; torch.cuda.synchronize()
    start = time.time(); _ = Xg @ Xg; torch.cuda.synchronize()
    print(f"GPU：{time.time() - start:.4f}秒")
````

<details><summary>ヒント / 解答と解説を見る</summary>

- **ヒント 1**：「Pythonのfor文 vs PyTorchテンソル演算」と「CPU vs GPU の行列積」の2段階に分けて考えましょう。
- **ヒント 2**：① テンソルは内部の高速なネイティブコードで一括処理する点、② 行列積の各出力マスは独立なので数千コアのGPUで同時並列計算できる点、の2点をまとめます。
- **解答の目安**：① PyTorchのテンソル演算はPythonの遅いfor文を1回ずつ解釈せず、内部のC/C++などで最適化された高速プログラムで全要素をまとめて処理するため速い。② 行列の掛け算の各マス $(i, j)$ の「掛けて足す」計算はお互いに独立しているため、少数の高性能コアで順番にこなすCPUよりも、数千個のコアで単純計算を同時に並列実行できるGPUの方が圧倒的に速い。

**考え方**：LLMの計算の大部分は巨大な行列の掛け算です。だからこそ、テンソルで並列化できる形に数式を書き直し、GPUで一気に計算する設計が現代AIの土台になっています。

1. Pythonの `for` 文 vs テンソル：1要素ごとのインタプリタ処理を避け、最適化済みネイティブコードで一括実行。
1. CPU vs GPU：行列積の各出力要素は互いに依存しないため、数千の単純コアで同時並列計算するGPUに完璧にマッチする。
