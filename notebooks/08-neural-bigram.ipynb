````python
import re, math, urllib.request
import torch
import torch.nn.functional as F

def load_soseki(file_id="773_ruby_5968"):
    url = ("https://raw.githubusercontent.com/aozorahack/aozorabunko_text/0984f7dc18a2270bdfd2ff99bf14fd1b3ebb4f65/"
           f"cards/000148/files/{file_id}/{file_id}.txt")
    try:
        with urllib.request.urlopen(url, timeout=30) as res:
            text = res.read().decode("cp932")
    except Exception as e:
        raise RuntimeError(f"青空文庫の取得に失敗しました（時間をおいて再実行するか付録Eを参照）: {e}") from e
    text = text.split("-------------------------------------------------------")[2]
    text = text.split("底本：")[0]
    text = re.sub(r"《.*?》", "", text)
    text = re.sub(r"※?［＃.*?］", "", text)
    text = text.replace("｜", "").replace("\r\n", "\n")
    text = re.sub(r"\n{3,}", "\n\n", text)
    return text.strip()

text = load_soseki()
chars = sorted(set(text))
V = len(chars)
stoi = {ch: i for i, ch in enumerate(chars)}
itos = {i: ch for ch, i in stoi.items()}

data = torch.tensor([stoi[c] for c in text])
n = int(len(data) * 0.9)
train_data, val_data = data[:n], data[n:]
print(V, train_data.shape, val_data.shape)
````

> **💡 理論のポイント：ラベルシフトと「教師強制（teacher forcing）」**
>
> `xs = train_data[:-1]`、`ys = train_data[1:]` のように、同じ文章を1文字だけ右にずらして入力と正解の組を作る手法を **ラベルシフト** と呼びます。そして学習時に、モデル自身の前の予測結果ではなく**毎回「本当の正解の文字」を入力として与えて次を予測させる**方法を **教師強制**（teacher forcing）と呼びます。
>
> - **なぜ教師強制を使うのか？**：学習初期のモデルはでたらめな文字ばかり出すため、自分の出力を次の入力にすると連鎖的に崩壊して学習が進みません。正解を毎回教えながら練習させることで、全位置の損失を安定かつ並列に計算できます。
> - **生成（推論）との違い**：第3章の自己回帰生成では正解が存在しないため、モデル自身の予測を次の入力へ戻します。この「学習時は正解入力、本番は自己出力」というズレは研究でも重要なテーマです（第11章の複数位置同時学習でもそのまま使います）。
>
> ※ なお、ここでは第5章と同様に簡潔さのため全テキスト `text` から語彙 `chars` を作っていますが、実務・研究では検証データからの情報漏れを防ぐため訓練データのみで語彙を構築します。

第6章で見たとおり、文字番号 `x` を one-hot ベクトル（`x` 番目だけ `1`、残り2060個は `0`）にして行列 `W` を右から掛けると、ちょうど `W` の `x` 行目がそのまま取り出されます。

```python no-run
# 数学的にはこの2つは完全に同じ計算！
# 方法A（第6章の行列積）：巨大な one-hot 行列 [B, 2061] を作って W と掛ける（0の掛け算が多くて遅い）
#   logits = F.one_hot(x, num_classes=V).float() @ W
# 方法B（この章の書き方）：直接 x 行目を取り出すだけ（同じ結果で圧倒的に速く、自動微分もそのまま動く）
#   logits = W[x]
```

そして初期値を全部 `0` にすると、すべての文字で $e^0 = 1$ になるため、softmax 後の確率はちょうど $\dfrac{1}{1 + 1 + \dots + 1} = \dfrac{1}{V}$ の完全な一様分布になります。

#### 📝 確認問題（`ch08-q02` · Lv.2 · W[x] の行取り出しと one-hot × W の数学的一致）

ニューラルbigramでは、形 `[V, V]` のロジットの表 `W` から `logits = W[x]` で次の文字の点数を取り出しています。これは第6章で学んだ行列の掛け算とどのような関係にありますか？

- **a.** `W[x]` のような整数インデックス参照は不連続な操作と見なされるため、PyTorch の自動微分（`backward()`）が動作せず勾配が一切流れない

- **d.** `x` を one-hot ベクトル（形 `[B, V]`）にして `one_hot(x) @ W` を計算するのと数学的に同じ結果になり、無駄な 0 の積を省いて高速に行を取り出せる

## 損失：交差エントロピー

第5章の損失（正解の確率の $-\log$ の平均）を、テンソルで計算してみます。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
def loss_by_hand(W, x, y):
    logits = W[x]
    probs = F.softmax(logits, dim=1)
    p_correct = probs[torch.arange(len(y)), y]   # 各行の「正解の文字」の確率
    return -p_correct.log().mean()

print("手作りの損失：", loss_by_hand(W, xs[:1000], ys[:1000]).item())
print("F.cross_entropy：", F.cross_entropy(W[xs[:1000]], ys[:1000]).item())
print("log V：", math.log(V))
````

```text

F.cross_entropy： 7.630943775177002
log V： 7.630946580890459
```

</details>

> **💡 理論のポイント：交差エントロピー：`F.cross_entropy` が1つに融合している3ステップ**
>
> `F.cross_entropy(logits, targets)` は、次の **3つの計算** を内部でひとまとめ（かつオーバーフロー・アンダーフローしない数値的に安全な方法）で実行してくれます。
>
> 1. **安全な softmax**：生の点数 `logits`（形 `[B, V]`）の各行から最大値を引いて確率分布 $p$ に変換（第3章）
> 2. **正解の驚き $-\log(p[\text{target}])$**：各サンプル $i$ について、正解文字 `targets[i]` の確率だけを取り出して $-\log$ をとる（第5章）
> 3. **バッチ平均**：$B$ 個のサンプルの平均をとって1つのスカラー（損失）を返す
>
> $$\text{Loss} = -\frac{1}{B} \sum_{i=1}^{B} \log \left( \text{softmax}(\text{logits}_i)_{\text{target}_i} \right)$$
>
> まっさらなモデルの損失は $\log V = 7.631$。第5章の記録表の「でたらめ」と同じです。ここから、どこまで下げられるでしょう？

#### ⚠️ つまずき注意：絶対に注意！ `F.cross_entropy` には「softmax前の生のロジット」を渡す

初心者が最もやりがちなバグが、**すでに `F.softmax` を掛けた確率 `probs` を `F.cross_entropy` に渡してしまう**ことです。`F.cross_entropy` の内部でもう一度 softmax が掛かってしまい（**二重softmax**）、エラーも出ないまま学習が壊れます！

```python no-run
import math
# 正解がクラス0で、モデルがロジット [5.0, 0.0, 0.0]（ほぼ確信！）を出した例：
# ① 正しい使い方（生のロジットを渡す）：
#    softmax([5, 0, 0]) ≒ [0.987, 0.007, 0.007] → 損失 = -log(0.987) ≒ 0.013（ほぼ完璧！）
# ② 誤った使い方（確率 [0.987, 0.007, 0.007] をさらに F.cross_entropy に渡す）：
#    もう一度 softmax されて [0.570, 0.215, 0.215] → 損失 = -log(0.570) ≒ 0.562（大外れ扱い！）
```

**鉄則**：モデルの最後は必ず生のロジットのまま返し、`F.cross_entropy(logits, y)` に直接渡しましょう。確率が欲しいとき（生成や上位表示）だけ `F.softmax` を掛けます。

> **🐍 Python・PyTorchのポイント：2つのリストで同時に選ぶ**
>
> `probs[torch.arange(len(y)), y]` は、「0行目の `y[0]` 列、1行目の `y[1]` 列、2行目の `y[2]` 列……」を一度に取り出す書き方です（第6章の `P[xs, ys]` と同じ）。`torch.arange(n)` は `0, 1, 2, …, n-1` を作ります。

#### 📝 確認問題（`ch08-q04` · Lv.2 · F.cross_entropy が融合する3処理と「生のロジットを渡す」鉄則）

`F.cross_entropy(logits, targets)` が内部でひとまとめに実行している3つの処理と、第1引数に渡すべきものの組み合わせとして**正しい**のはどれですか？

- **a.** ①予測値と正解の二乗誤差 → ②平方根 → ③バッチ合計。第1引数には必ず事前に `F.softmax` を掛けた確率を渡す

## 学習：ミニバッチで坂を下る

いよいよ学習です。14万個の問題を毎回全部解くと時間がかかるので、**毎回ランダムに1024問だけ**選んで損失を計算し、坂を下ります。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
g = torch.Generator().manual_seed(0)
W = torch.zeros((V, V), requires_grad=True)
lr = 50
history = []

for step in range(3001):
    ix = torch.randint(0, len(xs), (1024,), generator=g)   # ランダムに1024問選ぶ（(1024,) は形の指定。要素1つでもカンマが必要）
    loss = F.cross_entropy(W[xs[ix]], ys[ix])               # ① 損失
    W.grad = None                                            # （勾配をリセット）
    loss.backward()                                          # ② 勾配
    with torch.no_grad():
        W -= lr * W.grad                                     # ③ 更新
    if step % 250 == 0:
        with torch.no_grad():
            val_loss = F.cross_entropy(W[vxs], vys).item()
        history.append((step, loss.item(), val_loss))
        if step % 500 == 0:
            print(f"ステップ {step:4d}：訓練 {loss.item():.3f}  検証 {val_loss:.3f}")
````

> **🐍 Python・PyTorchのポイント：W.grad = None**
>
> 第7章では `.grad.zero_()` で勾配を0にしました。`W.grad = None` は「勾配を捨てる」で、効果は同じです（次の `backward()` で新しく作られます）。メモリの節約になるので、こちらがよく使われます。

<details><summary><b>基礎の補足：ニューラル学習ループ5ステップの解剖とミニバッチの読み方</b>（クリックで開く）</summary>

この先の章（MLP・GPT・ファインチューニング）でも、学習ループの心臓部は常に次の **5ステップ** です：

| 順番 | ステップ名 | この章のコード | 意味 |
|---|---|---|---|
| ① | **順伝播（forward）** | `logits = W[xs[ix]]` | ミニバッチ1024問の生の点数（形 `[1024, 2061]`）を出す |
| ② | **損失（loss）** | `loss = F.cross_entropy(logits, ys[ix])` | 正解とのずれを1つの数にまとめる |
| ③ | **勾配リセット（zero_grad）** | `W.grad = None` | 前回の勾配を消す（`backward()` の直前に置くのが安全！） |
| ④ | **逆伝播（backward）** | `loss.backward()` | 計算グラフを逆走し、全マスの勾配 `W.grad` を一括計算 |
| ⑤ | **更新（step）** | `with torch.no_grad(): W -= lr * W.grad` | 記録を切って勾配の逆向きへ一歩下る |

また `ix = torch.randint(0, len(xs), (1024,), generator=g)` のポイントも押さえておきましょう：
- `0` 以上 `len(xs)` 未満のランダムな行番号を `batch_size = 1024` 個引きます。
- Pythonでは要素1個のタプルには末尾カンマが必要なので、`(1024)` ではなく **`(1024,)`** と書きます（`(1024)` だとただの整数になりエラー）。
- 1エポック（訓練データ1周）＝ `len(xs) / batch_size` ステップなので、総エポック数は `(ステップ数 × batch_size) ÷ len(xs)` で計算できます。

</details>

> **💡 理論のポイント：ミニバッチ確率的勾配降下法（SGD）**
>
> 全部のデータで損失を計算する代わりに、ランダムに選んだ一部（**ミニバッチ**、今回は1024個）で計算して坂を下る方法を、**確率的勾配降下法**（stochastic gradient descent、**SGD**）と言います。
>
> - 一部だけなので、損失や勾配は少し**ぶれる**（上の「訓練」の値がガタガタなのはこのため）
> - でも計算が軽いので、同じ時間で**何倍も多く**坂を下れる
> - ぶれは、平均すると正しい方向を向いている
>
> 全問題を解いてから先生に採点してもらうより、10問ずつ解いてはその場で答え合わせをするほうが、早く上達しますよね。データ全体を1周することを **1エポック**（epoch）と呼びます。今回は1024問 × 3000回 ≒ 300万問なので、約21エポックです。
>
> 学習率が50と大きいのは、この表の1マスあたりの勾配がとても小さいからです。モデルごとに、ちょうどいい学習率は違います。

> **💡 理論のポイント：最尤推定：学習と数えるは同じ答えになる**
>
> これは偶然ではありません。「訓練データの交差エントロピー（驚きの平均）を最小にするような確率」は、数学的に **数えた割合（頻度）** とぴったり一致することが証明できます。この「データが最もありそうになる（尤もらしい）パラメータを選ぶ」という考え方を **最尤推定**（さいゆうすいてい、maximum likelihood estimation）と言います。
>
> なぜ一致するのかというと、交差エントロピー損失の傾き（勾配）を計算すると、ちょうど「モデルの予測確率 − 実際の出現割合」という形になるからです。予測確率が実際の頻度とぴったり一致した瞬間に、傾きがゼロになって坂の下りが止まります。だから勾配降下法を繰り返すだけで、自然と数えた割合と同じ確率に落ち着くのです。
>
> つまり、
>
> - 数えるbigram ＝ 最尤推定の答えを**直接計算**したもの
> - ニューラルbigram ＝ 同じ答えを**坂を下って探した**もの
>
> 直接計算できるのは、bigramのような単純なモデルだけです。ニューラルネットワークを使えば、直接計算できないような複雑なモデルでも、同じ「坂を下る」方法で答えを探せます。それが、この先のすべての章でやることです。

> **💡 理論のポイント：スムージングと正則化**
>
> 数えるbigramでは、見たことのないペアに確率0をつけないよう、スムージングで回数に少し下駄をはかせました。ニューラルbigramでは、`W` が全部0（＝全文字が等確率）から出発し、学習ステップ数が有限なので、極端な確率0にはなりません。
>
> さらに積極的に「極端な値になるのを防ぐ」方法として、損失に「`W` の値が大きいほど罰を与える」項を足すやり方があります。これを **正則化**（せいそくか、regularization）と言い、スムージングと同じ役割を果たします（第10章で `weight_decay` として登場します）。

ニューラルbigramを十分長く学習させると、softmax後の確率の表はどのような値に近づきますか？

- **a.** 訓練データで数えた実際の出現割合（頻度）に近づく（交差エントロピー最小化＝最尤推定の答えと一致するため）
- **b.** 勾配降下法はランダムなミニバッチを使うため、数えた出現割合とはまったく無関係な別解に収束する
- **c.** 各行で最もよく出る 1 文字だけが確率 `1.0` になり、他のすべての文字の確率は完全に `0.0` になる
- **d.** 学習が進むほどロジットの差が平均化され、どの文字もすべて等しく $1/V$ になる一様分布に近づく

1. 交差エントロピー損失のロジットに対する勾配 ＝ `予測確率 p − 実際の頻度 q`。

1. `p == q` で勾配が 0 になり、数えた頻度と同じ答えでピタリと止まる。

1. ロジットが 0 に近いほど確率は一様分布に近づくため、両者は同じ「極端さの抑制」を果たす。
