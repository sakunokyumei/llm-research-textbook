## 漱石の7作品を読み込む

第9章までと同じ `load_soseki`（下のセルで定義し直します）を使って、7作品をまとめて読み込みます。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
import re, math, urllib.request

def load_soseki(file_id="773_ruby_5968"):
    url = ("https://raw.githubusercontent.com/aozorahack/aozorabunko_text/0984f7dc18a2270bdfd2ff99bf14fd1b3ebb4f65/"
           f"cards/000148/files/{file_id}/{file_id}.txt")
    try:
        with urllib.request.urlopen(url, timeout=30) as res:
            text = res.read().decode("cp932")
    except Exception as e:
        raise RuntimeError(f"青空文庫の取得に失敗しました（時間をおいて再実行するか付録Eを参照）: {e}") from e
    text = text.split("-------------------------------------------------------")[2]
    text = text.split("底本：")[0]
    text = re.sub(r"《.*?》", "", text)
    text = re.sub(r"※?［＃.*?］", "", text)
    text = text.replace("｜", "").replace("\r\n", "\n")
    text = re.sub(r"\n{3,}", "\n\n", text)
    return text.strip()

works = {
    "こころ": "773_ruby_5968",
    "坊っちゃん": "752_ruby_2438",
    "吾輩は猫である": "789_ruby_5639",
    "三四郎": "794_ruby_4237",
    "草枕": "776_ruby_6020",
    "夢十夜": "799_ruby_6024",
    "それから": "56143_ruby_50824",
}
texts = {}
for title, file_id in works.items():
    texts[title] = load_soseki(file_id)
    print(f"{title}：{len(texts[title]):,} 文字")

all_text = "\n\n".join(texts.values())
chars = sorted(set(all_text))
V = len(chars)
stoi = {ch: i for i, ch in enumerate(chars)}
itos = {i: ch for ch, i in stoi.items()}
print(f"合計：{len(all_text):,} 文字、語彙：{V} 文字")
````

## 学習の道具をそろえる

ミニバッチを作る関数、検証損失を測る関数、学習率スケジュール（第10章）を用意します。検証損失は、検証データを128文字ずつに区切って**全部**で測るので、毎回同じ値になります。

🔎 **実行前に確認**：コードの各行が何を受け取り、何を作るか確認してから ▶ を押してください。

````python
block_size = 128    # 何文字前まで見るか（窓の大きさ）
batch_size = 32     # 1回に何個の窓を学習するか

def get_batch(data, g):
    ix = torch.randint(0, len(data) - block_size - 1, (batch_size,), generator=g)
    x = torch.stack([data[i:i + block_size] for i in ix])
    y = torch.stack([data[i + 1:i + block_size + 1] for i in ix])
    return x.to(device), y.to(device)

@torch.no_grad()
def val_loss(model, data):
    model.eval()
    starts = range(0, len(data) - block_size - 1, block_size)
    xs = torch.stack([data[i:i + block_size] for i in starts])
    ys = torch.stack([data[i + 1:i + block_size + 1] for i in starts])
    total = 0.0
    for j in range(0, len(xs), 64):
        _, loss = model(xs[j:j + 64].to(device), ys[j:j + 64].to(device))
        total += loss.item() * len(xs[j:j + 64])
    model.train()
    return total / len(xs)

def lr_at(step, max_lr, total, warmup=200):
    if step < warmup:
        return max_lr * step / warmup
    progress = (step - warmup) / (total - warmup)
    return max_lr * 0.5 * (1 + math.cos(math.pi * progress))
````

`.to(device)` で、データをGPUのメモリに送っています。GPUで計算するには、**モデルとデータの両方**がGPUにいる必要があります。

<details><summary><b>基礎の補足：model.train() / model.eval() と torch.no_grad() の違い（なぜ両方必要？）</b>（クリックで開く）</summary>

`val_loss` 関数には **`@torch.no_grad()`**・**`model.eval()`**・**`model.train()`** の3つが登場します。初学者が最も混同しやすい2つのスイッチを整理しておきましょう。

| スイッチ | 何を切り替えるか | 忘れると何が起きるか |
|---|---|---|
| `model.eval()` / `model.train()` | **Dropout などの層の挙動**（評価中はランダムな遮断をオフにして全ユニットを使う） | `eval()` 忘れ → 検証損失がブレて悪化。末尾の `train()` 忘れ → 以後の訓練で Dropout が永遠にオフになり過学習！ |
| `@torch.no_grad()` / `with torch.no_grad():` | **微分用の計算グラフ記録**（逆伝播のための途中メモを保存しない） | 忘れ → 評価中も不要なメモがGPUメモリ（VRAM）を圧迫して遅くなる・メモリ不足の原因に |

この2つはまったく別のスイッチなので、検証や文章生成のときは**必ず両方セット**で使い、学習ループに戻る前に `model.train()` へ戻します。

また、`lr_at(step, max_lr, total, warmup=200)` は、最初の `warmup` ステップで `0 → max_lr` まで直線的に上げ（ウォームアップ）、その後は $\frac{1}{2}(1 + \cos(\pi \cdot \text{progress}))$ に従って滑らかに `max_lr → 0` へ下げる（コサイン減衰）関数です。たとえば `max_lr=0.001, total=1000, warmup=200` なら、`step=0` で `0.0`、`step=100` で `0.0005`、`step=200` で最大 `0.001`、`step=600`（減衰の中間点）で `0.0005`、`step=1000` で `0.0` になります。

#### 📝 確認問題（`ch13-q03` · Lv.2 · model.eval() / model.train() と torch.no_grad() の違いと両方必要な理由）

## 学習ループ（GPU対応版）

第10章の学習ループに、GPUで速く学習するための工夫（**混合精度**）を足したものです。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
def train(train_data, max_iters, lr=1e-3, seed=0, log_every=500):
    torch.manual_seed(seed)
    g = torch.Generator().manual_seed(seed)
    model = GPT(V, block_size=block_size, n_embd=128, n_head=4, n_layer=4, dropout=0.1).to(device)
    opt = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=0.1)
    scaler = torch.amp.GradScaler(device, enabled=(device == "cuda"))
    history = []
    for step in range(max_iters + 1):
        for group in opt.param_groups:
            group["lr"] = lr_at(step, lr, max_iters)
        x, y = get_batch(train_data, g)
        with torch.autocast(device_type=device, dtype=torch.float16, enabled=(device == "cuda")):
            _, loss = model(x, y)
        opt.zero_grad(set_to_none=True)              # 勾配を0にする代わりに捨てる（第8章の W.grad = None と同じ）
        scaler.scale(loss).backward()
        scaler.unscale_(opt)
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        scaler.step(opt)
        scaler.update()
        if step % log_every == 0:
            vl = val_loss(model, kokoro_val)
            history.append((step, loss.item(), vl))
            print(f"ステップ {step:5d}：訓練 {loss.item():.3f}  検証（こころ）{vl:.3f}")
    return model, history
````

#### 💡 理論のポイント：混合精度：数の細かさを半分にして速くする

コンピュータの小数（浮動小数点数）には、使うビット数によって種類があります。

| 形式 | ビット数 | 1個のバイト数 | 特徴 |
|---|---|---|---|
| float32（単精度） | 32 | 4 | PyTorchの標準。正確 |
| float16（半精度） | 16 | 2 | メモリ半分。T4などのGPUで特に速い。表せる範囲が狭い |
| bfloat16 | 16 | 2 | 表せる範囲はfloat32並み、細かさは粗い。新しいGPU向け |

`torch.autocast(…, dtype=torch.float16)` の中では、行列の掛け算などの重い計算を、自動的にfloat16で行います。パラメータそのもの（マスターウェイト）や、細かさが大事な計算（損失の合計など）はfloat32のまま。2つの精度を混ぜて使うので **混合精度**（mixed precision）と呼びます。T4ではfloat32の何倍も速く計算できます。

「なぜパラメータ自体も全部float16にしないの？」と思うかもしれません。理由は、パラメータの更新量（学習率 $\times$ 勾配）がごく小さな値（例えば `0.00001` など）になるからです。float16の小さな数にその微小な更新量を足そうとすると、桁落ちして更新が無視されてしまいます。そこで、パラメータの「原本（マスターコピー）」はfloat32で正確に保持し、計算速度が求められる重い行列の掛け算だけfloat16に任せる、という役割分担をしています。

float16は小さな数を表すのが苦手で、とても小さな勾配が0になってしまう（第5章のアンダーフロー！）ことがあります。そこで `GradScaler` が、損失を一時的に何千倍にもしてから `backward()` し、更新の前に元に戻します。勾配を「拡大鏡で見る」作戦です。

CPUのときは `enabled=False` なので、これらは何もしません。同じコードがCPUでもGPUでも動きます。

> **📘 基本情報とのつながり：浮動小数点数の形式**
>
> 基本情報で、浮動小数点数は「符号部・指数部・仮数部」でできていると習いましたね。float16は指数部5ビット・仮数部10ビット、bfloat16は指数部8ビット・仮数部7ビットです。指数部が多いほど**表せる範囲**が広く、仮数部が多いほど**細かさ**が高い。bfloat16は「範囲は広く、細かさは我慢」という、深層学習向けに考えられた形式です（Googleが考案しました）。第19章の量子化で、この話の続きをします。

<details><summary><b>基礎の補足：勾配クリッピング（clip_grad_norm_）はなぜ必要？ unscale_ との順番も確認</b>（クリックで開く）</summary>

学習ループの後半にある次の2行は、大規模LLMの学習で欠かさず入れられる「安全ベルト」です。

```python no-run
scaler.unscale_(opt)                                      # ① 先に拡大を元に戻す
torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)   # ② 元の大きさの勾配を上限 1.0 で切る
```

**なぜ勾配爆発が起きるのか？**
深いネットワークの逆伝播では、たまたま難しいミニバッチに当たった瞬間などに勾配が極端に大きくなる（**勾配爆発**）ことがあります。そのまま更新すると1ステップで重みが吹き飛び、損失が急上昇したり `NaN` になったりします。

**L2 ノルムクリッピングの仕組み**：全パラメータの勾配を1本のベクトル $\boldsymbol{g}$ とみなし、長さ（L2 ノルム）$\|\boldsymbol{g}\|_2 = \sqrt{\sum_i g_i^2}$ を計算します。これが上限 `max_norm = 1.0` を超えたときだけ、全勾配に同じ係数 $\frac{1.0}{\|\boldsymbol{g}\|_2}$ を掛けて縮めます。

- たとえば $\boldsymbol{g} = [3.0, 4.0]$ なら $\|\boldsymbol{g}\|_2 = \sqrt{9 + 16} = 5.0 > 1.0$ なので、`1.0 / 5.0 = 0.2` 倍して `[0.6, 0.8]` にします（長さはちょうど `1.0`、**向きは `3:4` のまま不変**！）。
- $\|\boldsymbol{g}\|_2 \le 1.0$ なら何もしません。

なお、`scaler.scale(loss).backward()` の直後は勾配が数千倍に拡大されたままなので、**必ず先に `scaler.unscale_(opt)` で元の大きさに戻してから** `clip_grad_norm_` を呼ぶのが鉄則です。

#### 📝 確認問題（`ch13-q06` · Lv.2 · 混合精度（autocast + float32原本）と GradScaler の役割）

GPU 学習で float16 の混合精度を使うとき、`torch.amp.GradScaler` が必要な理由はどれですか？

- **a.** テキストデータの語彙数 `V` を float16 のビット幅に収まるよう 16 語彙単位に丸め込み、埋め込み層のメモリ転送量を削減するため
- **b.** CPU だけで計算するときに起きる丸め誤差を補正するための機能であり、GPU（`cuda`）環境では `enabled=False` として自動的に無効化される
- **c.** GPU の計算能力を引き出すために学習率を毎ステップ自動的に 2 倍ずつ拡大していき、通常より短いステップ数で最良の検証損失へ収束させるため
- **d.** float16 は非常に小さな数を表せず勾配が 0 に潰れる（アンダーフロー）ことがあるため、損失を一時的に大きく拡大してから逆伝播し、更新前に元の大きさに戻すため

<details><summary>ヒント / 解答と解説を見る</summary>

- **ヒント 1**：float16 で表現できる最小の正の数は約 $6 \times 10^{-5}$ です。それより小さな勾配が発生すると何が起きてしまうでしょうか？

1. `scaler.scale(loss).backward()`：損失を定数倍して勾配のアンダーフローを防ぐ。
1. `scaler.unscale_(opt)`：拡大した勾配を元のスケールへ割り戻す。
1. `scaler.step(opt)` / `scaler.update()`：もし `inf/nan` が出たら更新を飛ばして倍率を自動調整する。

#### 📝 確認問題（`ch13-q07` · Lv.2 · float16 と bfloat16 のビット配分（指数部 vs 仮数部）の違い）

同じ16ビット（2バイト）の浮動小数点数である **float16**（指数部5ビット・仮数部10ビット）と **bfloat16**（指数部8ビット・仮数部7ビット）の違いとして正しいものはどれですか？

- **a.** bfloat16 は float32 と同じ指数部 8 ビットを持つため「表せる数の範囲」が float32 並みに広くオーバーフロー／アンダーフローしにくい代わりに、細かさ（仮数部）は float16 より粗い
- **b.** bfloat16 は指数部へ 8 ビットを割り当てる代わりに符号部（1 ビット）を省略した形式であるため、正の数と 0 しか表現できず負の重みを扱えない
- **c.** float16 のほうが指数部のビット数が多くて表せる数の最大〜最小範囲が広く、bfloat16 のほうが仮数部のビット数が多くて有効桁数の細かさに優れている
- **d.** bfloat16 は精度を落とさずに広い範囲を表現するために 32 ビット（4 バイト）の記憶領域を使う形式であり、float16 に比べて 2 倍の GPU メモリを消費する

<details><summary>ヒント / 解答と解説を見る</summary>

- **ヒント 1**：浮動小数点数では「指数部」と「仮数部」がそれぞれ「数の範囲」と「有効数字の細かさ」のどちらを決めているか思い出しましょう。
- **ヒント 2**：どちらも合計 16 ビット（2 バイト）ですが、bfloat16 は指数部を float32 と同じ 8 ビットにしたため範囲が非常に広く、代わりに仮数部が 7 ビットと粗くなります。
- **正解**：`a`

**考え方**：深層学習の学習では「有効数字の細かい精度」よりも「勾配や損失がオーバーフロー／アンダーフローしない広い範囲」のほうが重要なため、新しいGPUでは bfloat16 が標準になっています。

1. float16：符号1 + 指数5 + 仮数10 = 16 bit（細かいが範囲が狭い → `GradScaler` が必要）。
1. bfloat16：符号1 + 指数8 + 仮数7 = 16 bit（範囲が float32 並みに広い → 学習が安定）。

3. **`"chars": chars`**：文字とトークンIDの対応表（これが1文字でもずれると生成結果が文字化けする）

#### 📝 確認問題（`ch13-q14` · Lv.3 · 記録表の検証損失からパープレキシティ（何択の迷い）への換算）

7作品で学習したミニGPTの検証損失は **`2.635`** でした。これをパープレキシティ（$e^{\text{loss}}$＝「だいたい何択で迷っているか」）に換算すると約何択になりますか？（参考：$\ln 10 \approx 2.303$、$e^{0.332} \approx 1.394$、誤差 $\pm 0.3$ まで正解）

<details><summary>ヒント / 解答と解説を見る</summary>

- **ヒント 1**：パープレキシティは $\mathrm{PPL} = e^{\text{loss}}$ です。指数法則 $e^{a + b} = e^a \cdot e^b$ を使って `2.635` を2つの数に分解してみましょう。
- **ヒント 2**：$2.635 = 2.303 + 0.332$ と分けられるので、$e^{2.635} = e^{2.303} \times e^{0.332} \approx 10 \times 1.394$ を計算します。
- **正解**：`13.94` 択
