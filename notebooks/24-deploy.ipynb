書き出しの関数です。第19章のabsmax量子化を、重みの**行ごと**に行って、8ビットの整数にします。そして、JSONに入れられるように **base64**（バイト列を文字だけで表す方法）に変換します。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
import base64, json
import torch

def quantize_rows(w):
    w = w.float()
    if w.dim() == 1:                                   # バイアスやLayerNormはそのまま（float32）
        return {"shape": list(w.shape), "f32": base64.b64encode(w.numpy().tobytes()).decode()}
    scale = w.abs().amax(dim=1).clamp(min=1e-8) / 127  # 行ごとの縮尺
    q = torch.round(w / scale[:, None]).clamp(-127, 127).to(torch.int8)
    return {"shape": list(w.shape),
            "i8": base64.b64encode(q.numpy().tobytes()).decode(),
            "scale": base64.b64encode(scale.numpy().astype("float32").tobytes()).decode()}

ck = torch.load(ckpt, map_location="cpu", weights_only=True)   # 安全のためテンソルと基本データのみ読み込む（第17章）
weights = {name: quantize_rows(t) for name, t in ck["model"].items() if not name.endswith("tril")}
with open("soseki-gpt.json", "w", encoding="utf-8") as f:
    json.dump({"config": ck["config"], "chars": ck["chars"], "weights": weights}, f, ensure_ascii=False)
print("書き出したサイズ：", round(os.path.getsize("soseki-gpt.json") / 1e6, 1), "MB")
````

> **📘 基本情報とのつながり：base64**
>
> **base64** は、どんなバイト列も、英数字と `+` `/` の64種類の文字だけで表す方法です。3バイト（24ビット）を6ビットずつ4文字に分けて表すので、サイズは約1.33倍になります。メールに画像を添付するときにも使われています。基本情報の「ビットの切り分け」の練習問題そのものですね。

<details><summary><b>基礎の補足：重みの書き出しとサイズ圧縮のしくみ（PyTorchテンソル → 軽量JSON）</b>（クリックで開く）</summary>

なぜそのままJSONに書かず、`quantize_rows` でひと工夫しているのでしょうか？ 約175万パラメータのサイズを比べると一目瞭然です。

| 書き出し方法 | 1パラメータあたり | 175万個の目安サイズ | 特徴 |
|---|---|---|---|
| 素の小数リスト（`float` そのままJSON） | 約18文字（18 B） | **約 31 MB** | 桁数が長すぎてWeb配信には重すぎる |
| 小数3桁丸め（`round(x, 3)` のJSON） | 約6〜7文字（6.5 B） | **約 11 MB** | 実装は超簡単だが、まだやや大きい |
| 生の `float32` バイナリ（`.pt` 相当） | 4 B | **約 7.0 MB** | JSONに直接入れられない（base64化で約9.3 MB） |
| **行ごと `int8` 量子化 ＋ base64** | 1 B × 1.33 | **約 2.5 MB** | 精度を保ちつつ最小！スマホでも1秒未満で読める |

また、**1次元テンソル（バイアスやLayerNorm）は要素数がごく少なく精度に敏感なので `float32` のまま残し**、因果マスクの **`tril` は固定の下三角表で推論時に再現できるため除外** しています。

#### 📝 確認問題（`ch24-q01` · Lv.2 · 重みの行ごと8ビット量子化・1次元層そのまま・tril除外の理由）

漱石GPTをブラウザ用のJSONに書き出す `quantize_rows` 関数で、①1次元テンソル（バイアスやLayerNorm）は量子化せず `float32` のまま残し、②名前が `tril` で終わるバッファは書き出しから除外している理由はどれですか？

- **a.** 1次元層は要素数が少なく精度劣化を避けたいから。`tril` は固定の下三角マスクで推論時に再生成でき保存がムダだから
- **b.** `tril` バッファをJSONに含めて書き出すと文字コード表が上書きされ、モデルが日本語ではなく英語しか話せなくなるから
- **c.** LayerNormやバイアスなどの1次元テンソルは学習終了後の推論計算では一度も使われず、JSONに書かなくても同じ出力になるから
- **d.** JavaScriptの言語仕様では8ビット整数（`Int8Array`）を一切扱えず、すべての重みを32ビット小数で持つしかないから

<details><summary>ヒント / 解答と解説を見る</summary>

- **ヒント 1**：175万パラメータの容量の大部分を占めるのは「1次元ベクトル」と「2次元行列」のどちらか、また `tril` は学習で変化する重みだったかを振り返りましょう。
- **ヒント 2**：1次元層（LayerNorm・バイアス）は小さくて精度に敏感である点と、`tril` が固定パターンの因果マスクである点の2つを結びつけてください。
- **正解**：`a`

**考え方**：モデル書き出しの鉄則は「①大部分の容量を占める2次元重みだけを効率よく量子化し、②小さくて敏感なLayerNorm/バイアスは高精度のまま残し、③学習パラメータではない固定マスク（`tril`）は省く」ことです。

1. 2次元行列（例：`[128, 512]`）→ 行ごとのabsmax 8ビット量子化でサイズを約1/4に圧縮。
1. 1次元ベクトル（例：`[128]`）→ 容量はほぼゼロなので `float32` のまま精度を保つ。
1. `tril` → 固定の0/1三角行列なのでJSONに書かず推論コード側で処理する。

#### 📝 確認問題（`ch24-q02` · Lv.3 · float32・int8・base64のファイルサイズ計算）

パラメータ数がちょうど **200万個**（$2 \times 10^6$ 個）のモデルがあるとします。全パラメータを `float32`（1個4バイト）でそのまま保存すると **8.0 MB** になりますが、これを `int8`（1個1バイト）に量子化し、さらにJSONに埋め込むため **base64**（サイズが約 $4/3 \approx 1.333$ 倍）に変換すると、重み本体のサイズはおよそ何 MB になりますか？（小数第1位まで、誤差 $\pm 0.2$ MB まで正解）

<details><summary>ヒント / 解答と解説を見る</summary>

- **ヒント 1**：まず全パラメータを `int8`（1個1バイト）に圧縮したときの生バイト数をMB単位で求め、そこにbase64の倍率を掛けます。
- **ヒント 2**：200万個 × 1バイト ＝ `2.0 MB` です。これにbase64変換の倍率 $4/3 \approx 1.333$ を掛けて小数第1〜2位まで求めましょう。
- **正解**：`2.67` MB

**考え方**：float32（8.0 MB）→ int8（2.0 MB、1/4に圧縮）→ base64（2.0 × 4/3 ≈ 2.67 MB）。base64で1.33倍になっても、元のfloat32の約1/3まで軽くできています！

1. `int8` の生バイト数：$2\times 10^6 \times 1\text{ バイト} = 2.0\text{ MB}$。
1. base64は3バイトを4文字（4バイト）で表すので、サイズは $\dfrac{4}{3} \approx 1.333$ 倍になる。
1. $2.0 \times 1.333 \approx 2.67\text{ MB}$。

### minigpt.js の中身を読む

> **🐍 Python・PyTorchのポイント：JavaScriptをPythonの目で読む**
>
> - `const` / `let`：変数を作る（`const` は後で入れ替えない変数）
> - `function f(x) { … return y; }`：`def f(x): … return y`
> - `for (let i = 0; i < n; i++) { … }`：`for i in range(n):`
> - `arr.map((v) => v * 2)`：`[v * 2 for v in arr]`
> - `Float32Array`：float32の数だけが並んだ配列（PyTorchのテンソルの、とても素朴な版）
> - `async` / `await`：「時間のかかる処理（ファイルの読み込みなど）を待つ」書き方
>
> インデント（字下げ）の代わりに `{ }` で範囲を表す、行の最後に `;` を付ける、くらいの違いで、考え方はPythonと同じです。

> **💡 理論のポイント：推論だけなら、どこでも動く（オンデバイス推論の魅力）**
>
> 学習には、自動微分（第7章）やAdam（第10章）、強力なGPUが必要でした。でも、**学習済みの重みで推論するだけ**なら、必要なのは掛け算と足し算と、exp・平方根などの簡単な関数だけです。だから、JavaScriptでも、スマートフォンでも、マイコンでも、原理的にはどこでも動かせます。
>
> 一般的なWebサービス（ChatGPTなど）は、ユーザーが文字を送ると、巨大なデータセンターのGPUサーバーへ通信が飛び、サーバー側で推論を計算して結果を返します。この方法だと、利用者が増えるほど高額なサーバー代（クラウド代）がかかります。
>
> それに対して、この章で作る方式は**オンデバイス推論（エッジ推論）**です。Webブラウザがモデルファイル（2.5MB）を最初に1回だけダウンロードし、そのあとは**アクセスした人のスマホやPCの頭脳（CPU）だけで推論を完結**させます。これには絶大なメリットがあります。
>
> - **サーバー代がほぼ0円**：静的ファイルを配信するだけなので、GitHub Pagesなどの無料サーバーで通常は料金がかかりません（ただしGitHub Pagesには月100GB程度の通信量目安があり、2.5MBのモデルならキャッシュなしで約4万回アクセス相当が目安です）
> - **入力が外に出ない（完全なプライバシー）**：ユーザーが打ち込んだ文章は、手元のブラウザの中だけで処理され、どこかの外部サーバーに送信されることがありません
> - **オフラインでも動く**：一度読み込んでしまえば、飛行機の中や電波のない地下鉄でも文章を生成できます
>
> ただし、数十億パラメータの大きなモデルは、端末でのダウンロードも計算も重くなります。175万パラメータ（2.5MB）くらいが、普通のブラウザで誰でも気軽に即座に試せる絶妙なサイズ感です。

````python
!pip install -q gradio
import gradio as gr
from transformers import pipeline

bot = pipeline("text-generation", model="Qwen/Qwen2.5-0.5B-Instruct", device_map="auto")

def text_of(content):
    if isinstance(content, list):          # 新しいGradioでは、文字以外の部品のリストのこともある
        return "".join(c.get("text", "") for c in content if isinstance(c, dict))
    return content

def reply(message, history):
    messages = [{"role": h["role"], "content": text_of(h["content"])} for h in history]
    messages.append({"role": "user", "content": text_of(message)})
    out = bot(messages, max_new_tokens=200)
    return out[0]["generated_text"][-1]["content"]

demo = gr.ChatInterface(reply, title="ボンゴ先生に質問しよう")
print(reply("こんにちは！", []))
````

> **💡 理論のポイント：GGUFと、量子化の名前**
>
> OllamaやLM Studioなどのアプリは、**GGUF** という形式のモデルファイルを使います。llama.cpp というC++で書かれた高速な推論プログラムのための形式で、量子化した重みをまとめて1つのファイルにしたものです。
>
> ファイル名についている `Q4_K_M` や `Q8_0` は、量子化の方法の名前です。`Q4` は約4ビット、`Q8` は約8ビット。数字が小さいほどファイルは小さく、性能は少し下がります。迷ったら `Q4_K_M` がよく使われる、バランスの良い選択です。第19章で手計算した量子化が、ここで使われているわけです。

#### 📝 確認問題（`ch24-q10` · Lv.3 · GGUF形式と量子化名（Q4_K_M / Q8_0）の読み方）

Ollamaやllama.cppで使われるGGUFモデルファイルで、同じ7Bモデルの `Q4_K_M` 版と `Q8_0` 版を比べたときの正しい説明はどれですか？

- **a.** `Q4_K_M` は圧縮テーブルの容量が大きいため、同じ7Bモデルでも常に `Q8_0` よりファイルサイズが2倍以上大きくなる
- **b.** GGUFはWebページの見た目やボタン配置を記述するためのHTML/CSS系のマークアップ言語で、重みデータは保存できない
- **c.** `Q4_K_M` は約4ビット量子化で軽量・高速なバランス型、`Q8_0` は約8ビット量子化でサイズは約2倍だが元の精度により近い
- **d.** `Q4_K_M` は総パラメータ数がちょうど4個だけの超小型モデルで、`Q8_0` はパラメータ数が8個に倍増した上位モデルである

<details><summary>ヒント / 解答と解説を見る</summary>

- **ヒント 1**：`Q4` と `Q8` の数字が「1パラメータあたりのビット数」を表していることを思い出しましょう。
- **ヒント 2**：ビット数が半分になるとファイルサイズやメモリ消費量、そして元の浮動小数点精度との近さがどう変わるか比べてください。
- **正解**：`c`

**考え方**：ローカルLLM選びでは、自分のPCのRAM/VRAM容量に合わせて量子化レベルを選びます。迷ったときの定番が容量と品質のバランスに優れた `Q4_K_M` です。

1. 7Bモデルの目安：FP16なら約14GB、`Q8_0`（約8bit）なら約7〜8GB、`Q4_K_M`（約4bit）なら約4〜5GB。
1. メモリに余裕があり精度重視 → `Q8_0`。
1. 一般PCで軽快に動かしたい → `Q4_K_M`。

#### 📝 確認問題（`ch24-q12` · Lv.5 · 【総合】用途・モデル規模・プライバシー・予算0円に応じた公開方式の設計）
