| GPT-2の名前 | 第12章の部品 | 意味 |
|---|---|---|
| `wte` | `tok_emb` | 文字（トークン）の埋め込み（word token embedding） |

違いは主に3つです。

1. **Q・K・Vを1つの全結合層でまとめて作る**（`c_attn` の出力が2304 = 768×3）。そして、ヘッドを別々のクラスにせず、1回の計算で全ヘッドをまとめて処理する

3. **出力層の重みを、トークンの埋め込み `wte` と共有している**（weight tying）。`state_dict` には `lm_head.weight` という名前もありますが、中身は `wte` と同じ表です（パラメータの数を数えるときは、1つとして数えられます）

さらに、GELUは計算の速い「tanhを使った近似版」が使われています。これらを合わせれば、読み込めるはずです。

#### 📝 確認問題（`ch15-q02` · Lv.2 · Q・K・Vをまとめた c_attn の出力次元数）

<details><summary>ヒント / 解答と解説を見る</summary>

- **ヒント 1**：`c_attn` は、Query・Key・Value の3種類のベクトルを1つの全結合層でまとめて出力します。
- **ヒント 2**：Q・K・V それぞれが幅 `n_embd = 768` なので、3本分の合計次元 $3 \times 768$ を計算しましょう。

**考え方**：Q・K・Vそれぞれ幅 `n_embd` のベクトルが必要なため、`c_attn` は `n_embd → 3 * n_embd` の線形変換になります。

1. `n_embd = 768`。
1. Q・K・Vの3つ分なので $3 \times 768 = 2{,}304$。
1. 計算後に `.split(768, dim=2)` で768ずつ3つに切り分ける。

## GPT-2と同じ形のGPTを書く

第12章のGPTを、上の3つの違いに合わせて書き直します。マルチヘッドAttentionは、ヘッドごとにクラスを作る代わりに、**形の組み替え**で全ヘッドを一度に計算します。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
class CausalSelfAttention(nn.Module):
    def __init__(self, n_embd, n_head, block_size):
        super().__init__()
        self.n_head = n_head
        self.c_attn = nn.Linear(n_embd, 3 * n_embd)     # Q・K・Vをまとめて作る
        self.c_proj = nn.Linear(n_embd, n_embd)
        self.register_buffer("tril", torch.tril(torch.ones(block_size, block_size)), persistent=False)

    def forward(self, x):
        B, T, C = x.shape
        hs = C // self.n_head                           # 1ヘッドの大きさ
        q, k, v = self.c_attn(x).split(C, dim=2)        # 最後の軸をC個ずつに切る（2304→768・768・768）。文字列のsplitとは別物
        q = q.view(B, T, self.n_head, hs).transpose(1, 2)   # [B, ヘッド, T, hs]
        k = k.view(B, T, self.n_head, hs).transpose(1, 2)
        v = v.view(B, T, self.n_head, hs).transpose(1, 2)
        att = q @ k.transpose(-2, -1) / math.sqrt(hs)   # 全ヘッドの点数を一度に
        att = att.masked_fill(self.tril[:T, :T] == 0, float("-inf"))
        att = F.softmax(att, dim=-1)
        y = att @ v                                     # [B, ヘッド, T, hs]
        y = y.transpose(1, 2).contiguous().view(B, T, C)    # ヘッドを横につなげ直す
        return self.c_proj(y)


class MLP(nn.Module):
    def __init__(self, n_embd):
        super().__init__()
        self.c_fc = nn.Linear(n_embd, 4 * n_embd)
        self.gelu = nn.GELU(approximate="tanh")         # GPT-2のGELU（近似版）
        self.c_proj = nn.Linear(4 * n_embd, n_embd)

    def forward(self, x):
        return self.c_proj(self.gelu(self.c_fc(x)))


class Block(nn.Module):
    def __init__(self, n_embd, n_head, block_size):
        super().__init__()
        self.ln_1 = nn.LayerNorm(n_embd)
        self.attn = CausalSelfAttention(n_embd, n_head, block_size)
        self.ln_2 = nn.LayerNorm(n_embd)
        self.mlp = MLP(n_embd)

    def forward(self, x):
        x = x + self.attn(self.ln_1(x))
        x = x + self.mlp(self.ln_2(x))
        return x


class GPT2(nn.Module):
    def __init__(self, vocab_size, block_size, n_embd, n_head, n_layer):
        super().__init__()
        self.block_size = block_size
        self.transformer = nn.ModuleDict(dict(
            wte=nn.Embedding(vocab_size, n_embd),
            wpe=nn.Embedding(block_size, n_embd),
            h=nn.ModuleList([Block(n_embd, n_head, block_size) for _ in range(n_layer)]),
            ln_f=nn.LayerNorm(n_embd),
        ))
        self.lm_head = nn.Linear(n_embd, vocab_size, bias=False)
        self.lm_head.weight = self.transformer.wte.weight    # 重み共有（weight tying）

    def forward(self, idx):
        B, T = idx.shape
        x = self.transformer.wte(idx) + self.transformer.wpe(torch.arange(T, device=idx.device))
        for block in self.transformer.h:
            x = block(x)
        return self.lm_head(self.transformer.ln_f(x))
````

> **💡 理論のポイント：ヘッドをまとめて計算する形の組み替え**
>
> 第12章では、4つのヘッドを、`Head` クラスから作った4つの部品で別々に計算していました。わかりやすいけれど、GPUにとっては「小さな仕事を何回も」頼まれるより「大きな仕事を1回でドカンと」頼まれるほうが、GPUコアを無駄なく使い切れてずっと高速なのです。
>
> Q・K・Vを作る3つの全結合層を `c_attn = nn.Linear(n_embd, 3 * n_embd)` の1つに合体させているのも同じ理由です。GPUに3回命令を出すオーバーヘッドを削り、1回の大きな行列積で計算してから、あとで3つに切り分けます。
>
> さらに、Q（形 `[B, T, 768]`）を `view` で `[B, T, 12, 64]`（12ヘッド×64）に切り分け、`transpose(1, 2)` で `[B, 12, T, 64]` に並べ替えます。こうすると、テンソルの `@` が「バッチごと・ヘッドごと」に行列の掛け算をしてくれるので、12ヘッド分の計算が**1回**で済みます。最後に逆の操作で `[B, T, 768]` に戻します（`contiguous()` は、並べ替えたテンソルをメモリ上できれいに並べ直すおまじない）。
>
> 計算の中身は、第11〜12章とまったく同じです。

> **🐍 Python・PyTorchのポイント：nn.ModuleDict と persistent=False**
>
> - `nn.ModuleDict(dict(wte=…, wpe=…))` は、部品に**名前をつけて**入れておく箱です。`dict(wte=1)` は `{"wte": 1}` と同じ意味の、辞書の別の書き方です。こうしておくと、部品の名前が `transformer.wte.weight` のようになり、GPT-2の重みの名前（上で見た一覧）と**一字一句同じ**になります。名前が同じなら、次の節で「同じ名前どうしをコピー」するだけで済みます
> - `register_buffer(…, persistent=False)` の `persistent=False` は、「`tril`（カンニング防止の三角形）は、保存する中身（`state_dict`）に入れない」という指定です。GPT-2の重みの一覧には `tril` がないので、これを付け忘れると、コピーのときに `KeyError: 'transformer.h.0.attn.tril'` になります

> **💡 理論のポイント：重み共有（weight tying）：同じ辞書の表と裏**
>
> 入力側の埋め込み `wte`（トークンID $\to$ 意味ベクトル）と、出力層 `lm_head`（意味ベクトル $\to$ 各トークンの点数）は、どちらも「語彙数 $\times$ 幅」の大きな行列です。
>
> これは英和辞典と和英辞典のようなものです。「単語から意味を引く」作業と、「意味から単語を引く」作業は表裏一体ですよね。GPT-2はこの2つを**まったく同じ表**にしています。「『猫』のベクトルに近いベクトルが計算されたら、次は『猫』の点数を高くする」という、入口と出口の対応関係を自然に表せるうえ、パラメータが3860万個（50257 $\times$ 768）も丸ごと節約できます。

#### 📝 確認問題（`ch15-q03` · Lv.3 · マルチヘッド一括計算のための view・transpose・contiguous）

自作GPT-2の `CausalSelfAttention` では、形 `[B, T, 768]` の `q` を `q.view(B, T, 12, 64).transpose(1, 2)` として形 **`[B, 12, T, 64]`** に並べ替えます。`transpose(1, 2)` で「ヘッド軸（12）」を「時間軸（T）」より前に持ってくる理由は何ですか？

- **a.** 入力文のトークン長 `T` をヘッド次元の64と入れ替えることで、あらゆる文章の長さを64トークンに固定するため
- **b.** PyTorchの仕様上、テンソルの軸1（2番目の軸）に12より大きな値を置くと行列積で実行時エラーになるから
- **c.** 後ろ2軸を `[T, 64]` にそろえ、全バッチ・全12ヘッドのスコア計算を1回の行列積でまとめて実行するため
- **d.** 12個のヘッドの特徴ベクトルを時間軸に沿って足し合わせ、計算量を減らすために1ヘッドへ統合してしまうため

<details><summary>ヒント / 解答と解説を見る</summary>

- **ヒント 1**：PyTorchのバッチ行列積演算子 `@` は、テンソルのどの軸どうしを掛け算するルールだったか思い出しましょう。
- **ヒント 2**：`@` は一番後ろの2軸を行列積し、それより前の軸を並列バッチとして扱います。`[B, 12, T, 64] @ [B, 12, 64, T]` の結果の形を確認しましょう。
- **正解**：`c`

## 重みを流し込む

同じ設定で自作GPT-2を作り、GPT-2の重みを1つずつコピーします。1つだけ注意点があります。GPT-2の公開コードでは、全結合層の重みを `[入力, 出力]` の向きで持っていますが、PyTorchの `nn.Linear` は `[出力, 入力]` の向きです。だから、該当する重みは **転置**（`.t()`、行と列の入れ替え）してからコピーします。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
mine = GPT2(cfg.vocab_size, cfg.n_positions, cfg.n_embd, cfg.n_head, cfg.n_layer)
mine.eval()
sd = mine.state_dict()
transposed = ("attn.c_attn.weight", "attn.c_proj.weight", "mlp.c_fc.weight", "mlp.c_proj.weight")

copied = 0
with torch.no_grad():
    for k in sd:
        src = sd_hf[k].t() if k.endswith(transposed) else sd_hf[k]
        assert sd[k].shape == src.shape, (k, sd[k].shape, src.shape)
        sd[k].copy_(src)
        copied += 1
print("コピーした重みの数：", copied)
````

`mine.state_dict()` が返す `sd` の中身は、コピーではなく、モデルの**パラメータそのもの**です。だから `sd[k].copy_(src)`（`src` の値で上書き）をすると、`mine` の中身が書き換わります。

`assert` は「この条件が正しくなければ、ここで止まって教えて」という命令です。形が1つでも合わなければエラーになるので、黙って間違ったコピーをすることはありません。

<details><summary><b>基礎の補足：なぜ転置 `.t()` が必要？ HFの `Conv1D` vs PyTorchの `nn.Linear`</b>（クリックで開く）</summary>

Hugging FaceのGPT-2実装は、OpenAIの元コード（TensorFlow由来）に合わせて、全結合層に **`Conv1D`** という独自クラスを使っています。

| 層の種類 | 計算式 | 重み行列の形（例：768 → 2304） | 転置 `.t()` は必要？ |
|---|---|---|---|
| HF GPT-2 の `Conv1D` | $x W + b$ | `[入力, 出力]` ＝ **`(768, 2304)`** | コピー元 |
| PyTorch の `nn.Linear` | $x W^\top + b$ | `[出力, 入力]` ＝ **`(2304, 768)`** | **必要！** |
| バイアス・LayerNorm・埋め込み | そのまま | `(2304,)`, `(768,)`, `(50257, 768)` | **不要** |

特に注意したいのは、正方形の `attn.c_proj.weight`（`(768, 768)`）です。縦横が同じなので、転置し忘れても `assert sd[k].shape == src.shape` を素通りしてしまいます。だからこそ、次の節で「出力ロジットが一致するか」を必ず検証します。

- **a.** PyTorchのテンソル型には完全一致を比較する `==` 演算子や `torch.equal` 関数が用意されていないから
- **b.** `torch.allclose` を使うと、一致しなかった要素の値を許容範囲内へ自動的に丸めて一致させてくれるから

## GPT-2に文章を書かせる

自作GPT-2で、文章を生成してみましょう。第3章の自作 `generate` と同じループに、上位k個だけから選ぶ **top-k** を加えたものです。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
@torch.no_grad()
def generate(model, text, n=40, temperature=0.8, top_k=50, seed=0):
    torch.manual_seed(seed)
    idx = tok(text, return_tensors="pt").input_ids
    for _ in range(n):
        logits = model(idx[:, -model.block_size:])[:, -1, :] / temperature
        v, _ = torch.topk(logits, top_k)
        logits[logits < v[:, [-1]]] = -float("inf")    # 上位k個以外は候補から外す
        next_id = torch.multinomial(F.softmax(logits, dim=-1), 1)
        idx = torch.cat([idx, next_id], dim=1)
    return tok.decode(idx[0])

print(generate(mine, "Hello, I'm a language model,"))
````

| 学習データ | 約100万文字 | 約40GB（数十億トークン） | 数兆〜数十兆トークン |
| 計算機 | Colabの無料GPU 1枚・数分 | 当時の大量のGPU・TPU | 数千〜数万枚のGPUで数か月 |

> **💡 理論のポイント：スケーリング則：大きくすると、予想どおりに賢くなる**
>
> 2020年、OpenAIの研究者たちは、LLMの損失が **モデルの大きさ $N$（パラメータ数）**、**データ量 $D$（トークン数）**、**計算量 $C$** を増やすと、とてもきれいな法則に従って下がることを見つけました。それぞれを10倍にすると、損失が決まった割合ずつ下がっていく（グラフの両軸を対数にすると直線になる、**べき乗則**）。これを **スケーリング則**（scaling law）と言います。
>
> この発見の何がすごいかというと、**小さな実験から、大きなモデルの性能が予想できる**ことです。「このくらいのお金をかければ、このくらい賢くなる」と計算できるので、巨大なモデルへの投資が始まりました。
>
> 2022年には、DeepMindの研究（通称 **Chinchilla**）が、「決まった**学習計算量**で一番賢いモデルを作るには、パラメータ1個あたり**約20トークン**のデータで学習させるのがよい」ことを示しました。それまでのモデルは、大きさのわりにデータが少なすぎた（読書量の割に頭だけ大きかった）のです。
>
> 一方、下の表の `Qwen2.5-0.5B` のように、最近の小さな公開LLMはChinchilla推奨（約98億トークン）の**約1,800倍**ものデータ（18兆トークン）で学習されています。これは**推論効率を優先した過剰学習（overtraining）**です。学習コストは増えても、完成後のモデルが小さいほど何億回もの推論（サービス提供や手元実行）が圧倒的に速く安くなるため、「小さく作って、とことん長く読ませる」設計が主流になっています。

計算量と、学習に必要なメモリを見積もってみましょう。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
def train_flops(n_params, n_tokens):
    return 6 * n_params * n_tokens          # 学習の計算量 ≈ 6 × N × D

def train_memory_gb(n_params):
    return n_params * 16 / 1e9              # AdamWで学習するときのメモリの目安（1パラメータ16バイト）

models = {"漱石GPT": (1.75e6, 2e7), "GPT-2 small": (124e6, 9e9),
          "Qwen2.5-0.5B": (0.49e9, 18e12), "70Bのモデル": (70e9, 15e12)}
for name, (N, D) in models.items():
    print(f"{name:12s} 計算量 {train_flops(N, D):9.1e} 回  学習メモリ {train_memory_gb(N):9.2f} GB"
          f"  Chinchilla推奨データ {20 * N:8.1e} トークン")
````

> **💡 理論のポイント：6ND と 16バイト**
>
> - **計算量 ≈ $6ND$**：1トークンを学習するごとに、パラメータ1個あたり約6回の掛け算・足し算をします（順方向で約2回、逆伝播で約4回）。T4の実効性能を毎秒10兆回（$10^{13}$）とすると、漱石GPTは数十秒、GPT-2 smallは約1週間、Qwen2.5-0.5Bは約170年（！）。大きなLLMの事前学習が、個人では無理な理由です
> - **メモリ ≈ 16バイト × パラメータ数（＋活性化）**：混合精度でAdamWを使うと、パラメータ1個あたり、重み（2バイト＋4バイトの控え）、勾配（2バイト）、Adamの慣性と歩幅の記録（4バイト×2）で約16バイト必要です。さらに実際の学習では、順伝播の途中結果である**活性化（第16章で詳しく）**のメモリもバッチサイズと文章長に応じて上乗せされます。T4のメモリは約15GBなので、全パラメータを学習できるのは**数億パラメータ**くらいまで。それより大きいモデルを無料GPUで鍛えるには、第19章の **LoRA** という技が必要になります

<details><summary><b>基礎の補足：スケーリング則の数式 $C \approx 6ND$ を1つずつ分解して手計算する</b>（クリックで開く）</summary>

記号の意味を並べると、驚くほどシンプルです。

- **$N$**：モデルのパラメータ数（例：1億 $= 10^8$）
- **$D$**：学習に使う総トークン数（例：20億 $= 2 \times 10^9$）
- **$C$**：学習全体で行う浮動小数点演算の回数（**FLOPs**）

**なぜ係数が「6」になるのか？**
行列の掛け算 $y = x W^\top$ では、1トークンに対して1個の重み $w$ が関わるたびに「掛け算1回＋足し算1回＝**2 FLOPs**」使います。

| フェーズ | 何を計算しているか | 1パラメータ・1トークンあたりの演算量 |
|---|---|---|
| 順伝播（forward） | 次の層の出力 $y = x W^\top$ | **2 FLOPs**（掛け算1＋足し算1） |
| 逆伝播①（backward） | 前の層へ伝える入力の勾配 $\partial L / \partial x$ | **2 FLOPs** |
| 逆伝播②（backward） | 自分の重みを更新する勾配 $\partial L / \partial W$ | **2 FLOPs** |
| **学習1ステップ合計** | **順伝播 ＋ 逆伝播** | **$2 + 4 = 6$ FLOPs** |

**具体的な手計算**：パラメータ数 $N = 10^8$（1億）のモデルを、Chinchillaの目安どおり $D = 20N = 2 \times 10^9$（20億）トークンで学習すると、

$$C \approx 6 \times 10^8 \times (2 \times 10^9) = 12 \times 10^{17} = 1.2 \times 10^{18}\ \text{FLOPs}$$

さらに1パラメータあたりの学習メモリ16バイトの内訳は、**fp16重み（2B）＋fp32控え重み（4B）＋勾配（2B）＋AdamWの1次・2次モーメント（4B＋4B）＝16バイト** なので、1億パラメータなら $10^8 \times 16 / 10^9 = 1.6$ GB になります（実運用ではここにバッチサイズと系列長に応じた活性化メモリが加算されます）。

</details>

#### 📝 確認問題（`ch15-q10` · Lv.3 · 学習計算量 C ≈ 6ND の「6」の理由（順伝播2＋逆伝播4））

LLMの事前学習にかかる総計算量（FLOPs）の近似式 **$C \approx 6ND$**（$N$：パラメータ数、$D$：学習トークン数）で、先頭の係数「**6**」はどのような内訳から来ていますか？

- **a.** AdamWオプティマイザが1パラメータの更新ごとに重みとモーメントで合計6バイトのメモリを読み書きするから
- **b.** 1トークン・1パラメータあたり、順伝播で約2 FLOPs、逆伝播で約4 FLOPs（入力勾配2＋重み勾配2）の計6 FLOPs
- **c.** 事前学習では同じコーパスを最低6エポック繰り返して学習させることが経験則として決まっているから
- **d.** 標準的なTransformer言語モデルが最小構成として必ず6層のブロックを積む設計になっているから

<details><summary>ヒント / 解答と解説を見る</summary>

- **ヒント 1**：行列の積和演算（掛けて足す）は、1パラメータあたり「掛け算1回＋足し算1回」で何FLOPsになるか考えましょう。
- **ヒント 2**：順伝播の行列積1回に加えて、逆伝播では「前の層へ渡す勾配」と「自分の重みの勾配」の2種類の行列積が必要です。
- **正解**：`b`

**考え方**：線形層の計算では、順伝播で $y = x W^\top$（2 FLOPs/param）、逆伝播で「前の層へ渡す勾配 $\partial L / \partial x$（2 FLOPs）」と「自分の重みの勾配 $\partial L / \partial W$（2 FLOPs）」の2種類の行列積を行うため、合計 $2 + 4 = 6$ FLOPsになります。

1. 順伝播：各パラメータ $w$ につき積和（掛け算1＋足し算1）＝ **2 FLOPs**。
1. 逆伝播①：前の層へ伝える活性化の勾配計算 ＝ **2 FLOPs**。
1. 逆伝播②：重み自身の更新用勾配計算 ＝ **2 FLOPs**。
1. 合計：$2 + 2 + 2 = 6$ FLOPs（推論だけなら順伝播のみなので約 $2ND$）。
