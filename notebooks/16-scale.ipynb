第15章のGPT-2形式のAttentionに、**KVキャッシュ**という仕組みを追加したものを用意します。重みはでたらめでかまいません（速さと正しさを確かめるだけなので）。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
import math, time
import torch
import torch.nn as nn
import torch.nn.functional as F

class CachedAttention(nn.Module):
    def __init__(self, n_embd, n_head):
        super().__init__()
        self.n_head = n_head
        self.c_attn = nn.Linear(n_embd, 3 * n_embd)
        self.c_proj = nn.Linear(n_embd, n_embd)

    def forward(self, x, cache=None):
        B, T, C = x.shape
        hs = C // self.n_head
        q, k, v = self.c_attn(x).split(C, dim=2)
        q, k, v = [t.view(B, T, self.n_head, hs).transpose(1, 2) for t in (q, k, v)]
        if cache is not None:                      # 過去のKとVを使い回す
            if "k" in cache:
                k = torch.cat([cache["k"], k], dim=2)
                v = torch.cat([cache["v"], v], dim=2)
            cache["k"], cache["v"] = k, v
        L = k.shape[2]                             # 過去も含めた長さ
        att = q @ k.transpose(-2, -1) / math.sqrt(hs)
        mask = torch.tril(torch.ones(L, L, device=x.device))[L - T:]    # 新しいT個の行だけのマスク
        att = att.masked_fill(mask == 0, float("-inf"))
        y = F.softmax(att, dim=-1) @ v
        return self.c_proj(y.transpose(1, 2).contiguous().view(B, T, C))

class TinyGPT(nn.Module):
    def __init__(self, V=1000, n_embd=256, n_head=4, n_layer=6, block_size=512):
        super().__init__()
        self.wte, self.wpe = nn.Embedding(V, n_embd), nn.Embedding(block_size, n_embd)
        self.ln1 = nn.ModuleList([nn.LayerNorm(n_embd) for _ in range(n_layer)])
        self.attn = nn.ModuleList([CachedAttention(n_embd, n_head) for _ in range(n_layer)])
        self.ln2 = nn.ModuleList([nn.LayerNorm(n_embd) for _ in range(n_layer)])
        self.mlp = nn.ModuleList([nn.Sequential(nn.Linear(n_embd, 4 * n_embd), nn.GELU(), nn.Linear(4 * n_embd, n_embd)) for _ in range(n_layer)])
        self.ln_f, self.head = nn.LayerNorm(n_embd), nn.Linear(n_embd, V)

    def forward(self, idx, caches=None, start_pos=0):
        T = idx.shape[1]
        x = self.wte(idx) + self.wpe(torch.arange(start_pos, start_pos + T, device=idx.device))
        for i in range(len(self.attn)):
            x = x + self.attn[i](self.ln1[i](x), None if caches is None else caches[i])
            x = x + self.mlp[i](self.ln2[i](x))
        return self.head(self.ln_f(x))

torch.manual_seed(0)
model = TinyGPT().eval()
print(f"パラメータ：{sum(p.numel() for p in model.parameters()):,}")
````

## ムダに気づく：生成のたびに全部計算し直している

第3章からずっと使ってきた生成ループを思い出してください。

> 文章全体をモデルに通す → 最後の位置の確率で1トークン選ぶ → 付け足す → **また文章全体をモデルに通す** …

100トークン目を生成するとき、1〜99トークン目の計算を、またゼロからやり直しています。でも、因果マスクのおかげで、**過去の位置の計算結果は、未来のトークンが増えても変わりません**（第11章のカンニング検査で確かめましたね）。だったら、前回の計算結果を取っておいて使い回せばいいはずです。

> **💡 理論のポイント：KVキャッシュ**
>
> Attentionで、過去の位置から受け取るのは **Key** と **Value** でした。過去の位置のKeyとValueは、一度計算したら変わりません。そこで、
>
> - 過去のトークンのK・Vを、層ごとに **キャッシュ**（取っておく場所）に保存しておく
> - 新しいトークンが来たら、**その1トークンぶんだけ**Q・K・Vを計算し、Kと Vをキャッシュの後ろに付け足す
> - 新しいトークンのQで、キャッシュの全K・Vに対してAttentionを計算する
>
> こうすると、1トークンを生成するための計算が「文章全体」から「新しく生まれた1トークンぶん」に減ります。もしキャッシュがないと、$N$ 個のトークンを出すのに毎回 $1 + 2 + \dots + N \approx N^2/2$ トークンぶんの重複計算をしてしまいますが、キャッシュがあれば過去の計算をそのまま再利用できるため、生成が劇的に速くなります。上のコードの `cache` がそれです。これを **KVキャッシュ** と言い、ほぼすべてのLLMの推論で使われています。

キャッシュなし・ありの2通りで生成して、**結果が同じで、速さが違う**ことを確かめましょう。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
@torch.no_grad()
def generate_naive(model, idx, n):
    for _ in range(n):
        logits = model(idx)                              # 毎回、全体を計算
        idx = torch.cat([idx, logits[:, -1:].argmax(-1)], dim=1)
    return idx

@torch.no_grad()
def generate_cached(model, idx, n):
    caches = [{} for _ in model.attn]
    logits = model(idx, caches)                          # 最初だけ全体を計算
    pos = idx.shape[1]
    for _ in range(n):
        next_id = logits[:, -1:].argmax(-1)
        idx = torch.cat([idx, next_id], dim=1)
        logits = model(next_id, caches, start_pos=pos)   # 新しい1トークンだけ
        pos += 1
    return idx

prompt = torch.randint(0, 1000, (1, 20))
# time.perf_counter() は「今の時刻（秒）」。前後の差が、かかった時間になる
t0 = time.perf_counter(); a = generate_naive(model, prompt, 200); t1 = time.perf_counter()
b = generate_cached(model, prompt, 200); t2 = time.perf_counter()
print("結果は同じ？", torch.equal(a, b))
print(f"キャッシュなし {t1 - t0:.2f}秒 / キャッシュあり {t2 - t1:.2f}秒 → {(t1 - t0) / (t2 - t1):.1f}倍速い")
````

#### 📝 確認問題（`ch16-q01` · Lv.2 · 因果マスクによりKVキャッシュが数学的に一致し高速化できる理由）

KVキャッシュを使うと、素朴な生成ループと**まったく同じ出力トークン**を出しながら何倍も速くなります。過去のトークンのKeyとValueを再計算せずに使い回しても結果が変わらない根拠は何ですか？

- **a.** 過去のK・Vは新トークンごとに少しずつ変化するが、誤差が小さいため無視して使い回す近似手法だから
- **b.** 推論ループのたびにTransformerの重みを再初期化して、過去のK・Vが毎回同じ基準値へ戻る仕組みだから
- **c.** 新しく生成されたトークンのAttention計算では、過去のトークン列を一切参照せず独立に処理するから
- **d.** 因果マスクにより過去トークンは未来を参照しないため、後ろに新トークンが増えても過去のK・Vが不変だから

<details><summary>ヒント / 解答と解説を見る</summary>

- **ヒント 1**：第11章で行った「後ろのトークンを書き換えても、前の位置の計算結果が変わらない」カンニング検査を思い出しましょう。
- **ヒント 2**：因果マスク（下三角行列）があるとき、位置 $t$ の隠れ状態やKey・Valueが位置 $t+1$ 以降に依存するかどうかを考えてみましょう。
- **正解**：`d`

**考え方**：因果マスクがあるため、位置 $t$ の隠れ状態とK・Vは位置 $1 \dots t$ だけに依存します。したがって位置 $t+1$ 以降が追加されても過去のK・Vは不変であり、安全に使い回せます。

1. 因果マスク（下三角行列）により、過去トークンは未来トークンを参照できない。
1. よってトークンが増えても過去位置のKey・Valueは完全に不変。
1. 新トークン1個分だけQ・K・Vを計算してキャッシュに連結すれば、出力は完全一致しつつ重複計算 $1 + 2 + \dots + N \approx N^2/2$ を削れる。

#### 📝 確認問題（`ch16-q02` · Lv.3 · KVキャッシュで過去のQuery（Q）を保存しない理由）

KVキャッシュは「Key」と「Value」だけを保存し、過去の「**Query（Q）**」は保存しません。過去のQueryを保存しなくてよい理由は何ですか？

- **a.** QueryはKeyやValueより特徴次元が何倍も大きく設計されており、キャッシュするとGPUメモリに入り切らないから
- **b.** 因果マスクを適用するとAttentionの式からQueryの項が消え、KeyとValueの内積だけで次トークンが求まるから

<details><summary>ヒント / 解答と解説を見る</summary>

- **ヒント 2**：時刻 $t$ で新しく出力を求める式 $\text{softmax}(q_t K_{1:t}^\top / \sqrt{d}) V_{1:t}$ の中に、過去の $q_1, \dots, q_{t-1}$ が現れるか確認しましょう。

1. Attentionは「Queryが過去のKeyと照合し、Valueを取り出す」仕組み。
1. ステップ $t$ で新しい出力が必要なのは位置 $t$ だけ → 必要なのは $q_t$ のみ。
1. 照合対象は過去全部なので $K_{1:t}$ と $V_{1:t}$ は必要だが、過去の $q_1 \dots q_{t-1}$ は二度と使わない。

## KVキャッシュのメモリと、GQA

キャッシュは速さと引き換えに **メモリ** を使います。どれくらい使うのか、計算してみましょう。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
def kv_cache_bytes(n_layer, n_kv_head, head_dim, n_tokens, bytes_per_num=2):
    return 2 * n_layer * n_kv_head * head_dim * n_tokens * bytes_per_num   # 2 = KとV

# Qwen2.5-0.5B：24層、Queryのヘッド14、K・Vのヘッド2、ヘッドの大きさ64
per_token = kv_cache_bytes(24, 2, 64, 1)
print(f"1トークンあたり {per_token / 1024:.0f} KB → 32,768トークンで {kv_cache_bytes(24, 2, 64, 32768) / 1e9:.2f} GB")
print(f"もしK・Vのヘッドも14個なら → 32,768トークンで {kv_cache_bytes(24, 14, 64, 32768) / 1e9:.2f} GB")
````

> **💡 理論のポイント：GQA：Key・Valueのヘッドを共有する**
>
> Qwen2.5-0.5Bは、Queryのヘッドが14個あるのに、KeyとValueのヘッドは**2個だけ**です。7個のQueryヘッドが、1組のKey・Valueを**共有**しているのです。これを **GQA**（Grouped-Query Attention、グループ化クエリAttention）と言います。
>
> 性能はほとんど落ちずに、KVキャッシュのメモリが7分の1になります（上の計算では2.82GB → 0.40GB）。長い文章を扱うLLMや、たくさんの人が同時に使うサービスでは、KVキャッシュのメモリが一番の悩みの種なので、最近のLLMのほとんどがGQAを採用しています。K・Vのヘッドを1組だけにした極端な版は **MQA**（Multi-Query Attention）と呼ばれます。

<details><summary><b>基礎の補足：数式の解剖：なぜKVキャッシュは $2 \times L \times H_{kv} \times d \times T \times b$ になるのか？</b>（クリックで開く）</summary>

キャッシュに保存するテンソルは、全 $L$ 層それぞれに「Key用とValue用の **2つ**」、形 `[H_kv, T, d]`、1数値あたり $b$ バイト（fp16なら $b=2$）です。したがって、

$$\text{KVキャッシュ（バイト）} = \underbrace{2}_{\text{KとV}} \times \underbrace{L}_{\text{層数}} \times \underbrace{H_{kv}}_{\text{KVヘッド数}} \times \underbrace{d}_{\text{ヘッド次元}} \times \underbrace{T}_{\text{トークン数}} \times \underbrace{b}_{\text{バイト数}}$$

| 方式 | Queryヘッド数 $H_q$ | KVヘッド数 $H_{kv}$ | KVキャッシュの大きさ | 特徴 |
|---|---|---|---|---|
| **MHA**（従来型） | 14 | 14 | 大きい（基準 1倍） | 品質は高いが長文でVRAMを圧迫 |

## 賢いサイコロ：top-k と top-p

第3章では、温度でサイコロの尖り具合を調整しました。でも、温度を上げると「ありえない候補」まで選ばれてしまう問題がありました（温度1.5の支離滅裂な文章を思い出してください）。そこで、**候補をあらかじめ絞ってから**サイコロを振る方法が使われます。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
words = ["猫", "犬", "人間", "名前", "机", "宇宙", "の"]
logits = torch.tensor([3.0, 1.6, 1.2, 0.4, -1.0, -1.5, -2.0])
probs = F.softmax(logits, dim=0)

def top_k_filter(probs, k):
    keep = torch.zeros_like(probs)
    idx = torch.topk(probs, k).indices
    keep[idx] = probs[idx]
    return keep / keep.sum()

def top_p_filter(probs, p):
    sorted_p, order = torch.sort(probs, descending=True)
    cum = torch.cumsum(sorted_p, dim=0)
    n_keep = int((cum < p).sum()) + 1            # 累積がpを超えるまでの最小の候補数
    keep = torch.zeros_like(probs)
    keep[order[:n_keep]] = probs[order[:n_keep]]
    return keep / keep.sum()

show = lambda p: "  ".join(f"{w}{x:.2f}" for w, x in zip(words, p) if x > 0)
print("そのまま ：", show(probs))
print("top-k=3 ：", show(top_k_filter(probs, 3)))
print("top-p=0.9：", show(top_p_filter(probs, 0.9)))
````

> **💡 理論のポイント：top-k と top-p（nucleus sampling）**
>
> - **top-k**：確率の高い上位 **k個** だけを機械的に候補に残す方法です。「テストの成績上位k人を常に合格にする」定員固定の選抜のようなものです。シンプルですが、1位の候補が確率99%で確定している場面でも無理やりk個の候補を残してしまい、たまに事故が起きます。
> - **top-p**（nucleus sampling、核サンプリング）：確率の高い順に足していき、合計が **p**（例えば0.9＝90%）を超えるまでの候補だけを残す方法です。「上位から足して得点全体の9割を占めるグループを選ぶ」方式です。モデルが答えに自信満々なときは1〜2個でスッと打ち切られ、迷っているときは自然に多くの候補が残るため、状況に応じて賢く柔軟に絞り込めます。
>
> どちらも、残した候補の確率を足して1になるように割り直してから、サイコロを振ります。実際のチャットでは、「温度0.7、top-p 0.8」のような組み合わせがよく使われます。
>
> もう1つ、同じ言葉をくり返しすぎるのを防ぐ **くり返しペナルティ**（repetition penalty）もよく使われます。すでに出てきたトークンの点数を少し下げる、という単純な方法です。第3章で小さなモデルが同じフレーズをくり返したのを、これで和らげられます。

## 学習の工学1：勾配累積

T4のメモリに、バッチ64が入らない。でもバッチ64で学習したい。そんなときの技が **勾配累積**（gradient accumulation）です。小さなバッチで何回か `backward()` して勾配を**ためて**から、まとめて1回更新します。第7章で「`.grad` は足し算されていく」と注意しましたが、それを逆に利用するのです。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
torch.manual_seed(0)
net = nn.Linear(16, 4)
x, y = torch.randn(32, 16), torch.randint(0, 4, (32,))

net.zero_grad()
F.cross_entropy(net(x), y).backward()            # バッチ32で1回
g_big = net.weight.grad.clone()

net.zero_grad()
for i in range(4):                               # バッチ8を4回に分ける
    xb, yb = x[i * 8:(i + 1) * 8], y[i * 8:(i + 1) * 8]
    (F.cross_entropy(net(xb), yb) / 4).backward()    # 4で割ってから足し込む
g_acc = net.weight.grad

print("勾配は同じ？", torch.allclose(g_big, g_acc, atol=1e-6))
````

<details><summary><b>📊 期待される出力（実測）</b>（実行後に自分の出力と見比べよう）</summary>

```text
勾配は同じ？ True
```

</details>

メモリに乗るのは「バッチ8」ぶんだけなのに、バッチ32で計算したのと同じ勾配が手に入りました。計算の量はバッチ32を1回やるのと同じなので、時間は大きくは変わりません（小分けにするぶん、少し遅くなる程度）。減るのは**途中の計算結果（活性化）のメモリ**で、だいたい4分の1になります。モデルの重み・勾配・AdamWの記録（第15章の「1パラメータあたり16バイト」）は、バッチの大きさに関係なく必要なので減りません。損失を累積回数（4）で割るのを忘れないのがコツです。

<details><summary><b>基礎の補足：なぜ「損失 ÷ `accum_steps`」が必須なのか？4サンプルで確かめる</b>（クリックで開く）</summary>

`F.cross_entropy` はバッチ内の**平均**を返します。4サンプルの損失 $\ell_1, \ell_2, \ell_3, \ell_4$ を「バッチ4一括」と「バッチ2 × 2回累積」で比べてみましょう。

$$\text{バッチ4一括の平均} = \frac{\ell_1 + \ell_2 + \ell_3 + \ell_4}{4} = \frac{1}{2}\left(\frac{\ell_1 + \ell_2}{2}\right) + \frac{1}{2}\left(\frac{\ell_3 + \ell_4}{2}\right)$$

それぞれのミニバッチ平均を `accum_steps`（ここでは `2`）で割ってから足し込むと、一括平均と**数学的に完全一致**します。もし割り忘れると勾配が `accum_steps` 倍に膨らみ、学習率を勝手に何倍にもしたのと同じになってしまいます。また、`opt.step()` と `opt.zero_grad()` は「`accum_steps` 回たまった後に1回だけ」呼びます。

</details>

#### 📝 確認問題（`ch16-q10` · Lv.3 · 勾配累積で損失を accum_steps で割る理由と zero_grad() のタイミング）

マイクロバッチサイズ `8` で `accum_steps = 4` 回の**勾配累積**を行い、実質バッチサイズ `32` の学習を再現します。もし誤って損失を `4` で割らずに `loss.backward()` を4回呼んでから `opt.step()` すると、何が起きますか？

- **a.** 毎回 `.grad` が新しい値で上書きされるため、最初の3回分が捨てられて4回目だけで更新されてしまう
- **b.** `.grad` に4回分が加算されて本来の4倍の大きさになり、学習率を4倍にしたのと同様に不安定化する
- **c.** PyTorchが累積回数を自動検出して平均化するため、バッチ32一括と数学的に完全に同じ勾配になる
- **d.** ミニバッチが4分割された影響で勾配の大きさが本来の4分の1に縮み、学習がほとんど進まなくなる

<details><summary>ヒント / 解答と解説を見る</summary>

- **ヒント 1**：PyTorchで `zero_grad()` を呼ばずに `backward()` を繰り返すと、`.grad` には勾配が加算されていきます。
- **ヒント 2**：8件ずつの「平均損失」を4回そのまま足し合わせると、32件全体の「平均損失」の何倍の大きさになるか計算してみましょう。
- **正解**：`b`

**考え方**：32件全体の平均損失は $\frac{1}{4}(\bar{L}_1 + \bar{L}_2 + \bar{L}_3 + \bar{L}_4)$ です。各マイクロバッチで $\bar{L}_i / 4$ にしてから `backward()` することで初めて、バッチ32一括と数学的に完全一致します。

1. バッチ32の平均損失 $= \frac{1}{32}\sum_{j=1}^{32} \ell_j$。
1. バッチ8の平均を4つに分けると $= \frac{1}{4}\left(\frac{1}{8}\sum_{1..8}\ell_j + \dots + \frac{1}{8}\sum_{25..32}\ell_j\right)$。
1. したがって各回 `loss / 4` を `backward()` し、4回たまった後に `opt.step()` と `opt.zero_grad()` を1回だけ呼ぶ。

1. 一括平均：`(2+4+6+8+10+12+14+16) / 8 = 72 / 8 = 9.0`。
1. 4つのチャンク平均：`3.0, 7.0, 11.0, 15.0`。
1. それぞれを4で割って足す：`(3.0 + 7.0 + 11.0 + 15.0) / 4 = 9.0` → `9.0 9.0 True`。

## 学習の工学2：活性化の再計算

学習中のメモリを一番食うのは、パラメータではなく、**逆伝播のために取っておく途中の計算結果**（活性化）であることがよくあります。第7章の計算グラフを思い出してください。`backward()` のために、順方向の途中の値を全部覚えておく必要がありました。

> **💡 理論のポイント：勾配チェックポイント**
>
> **勾配チェックポイント**（activation checkpointing）は、途中の値を**覚えておかずに捨ててしまい**、逆伝播で必要になったときに**もう一度計算し直す**方法です。計算時間が約3割増える代わりに、メモリを大幅に節約できます。「メモを取らずに、必要なときに考え直す」作戦です。PyTorchでは `torch.utils.checkpoint.checkpoint(block, x)` のように、ブロックを包むだけで使えます。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
from torch.utils.checkpoint import checkpoint

block = nn.Sequential(nn.Linear(64, 256), nn.GELU(), nn.Linear(256, 64))
x = torch.randn(8, 64, requires_grad=True)
out1 = block(x).sum(); out1.backward(); g1 = x.grad.clone()
x.grad = None
out2 = checkpoint(block, x, use_reentrant=False).sum(); out2.backward(); g2 = x.grad
print("同じ勾配？", torch.allclose(g1, g2))
````

<details><summary><b>基礎の補足：活性化の再計算：何を捨てて、何を残すのか？</b>（クリックで開く）</summary>

Transformerの24層すべての中間計算（LayerNorm後・Q/K/V・Attentionスコア・MLPの4倍幅の中間層など）をVRAMに残すと、長文では重み本体より何倍もメモリを食います。

| 方式 | 順伝播でVRAMに保存するもの | 逆伝播の動き | 計算時間 | 勾配の精度 |
|---|---|---|---|---|
| 通常学習 | 全ブロック内部の中間値すべて | 保存した値をそのまま読む | 基準（1.0倍） | 正確 |
| **勾配チェックポイント** | **各ブロックの入口 `x` だけ** | 必要な瞬間にそのブロックだけ再計算 | 約1.3倍（+3割） | **数学的に完全一致** |

推論時（`torch.no_grad()`）はそもそも逆伝播をしないため、勾配チェックポイントは**学習専用の省メモリ技**です。

#### 📝 確認問題（`ch16-q12` · Lv.2 · 活性化の再計算（勾配チェックポイント）のトレードオフ）

**勾配チェックポイント（activation checkpointing）** の仕組みとトレードオフとして正しいものはどれですか？

- **a.** モデルの重みテンソルを学習中にディスクへ退避させることで、推論時のトークン生成速度を5倍に高める技術
- **b.** 順伝播の活性化を捨て逆伝播で再計算することで、時間は約3割増えるが同じ勾配のままメモリを節約できる
- **c.** 学習途中のモデル重みを一定ステップごとにストレージへ保存しておく、障害復旧用チェックポイント機能のこと
- **d.** 途中のTransformer層の逆伝播を丸ごと省略して近似勾配を使うため、計算は2倍速くなるが精度が落ちる

<details><summary>ヒント / 解答と解説を見る</summary>

- **ヒント 1**：「メモを取って机を埋める代わりに、必要になったときに考え直す」というトレードオフを思い出しましょう。
- **ヒント 2**：中間値（活性化）をVRAMに残さず逆伝播で再計算すると、メモリ・計算時間・勾配の正確さはそれぞれどうなるか整理しましょう。
- **正解**：`b`

**考え方**：「計算時間（約1.3倍）」と引き換えに「活性化メモリ」を劇的に減らす、時間とメモリの典型的なトレードオフです。

1. 通常：全ブロックの途中の値をVRAMに保存 → 長文や深いモデルでVRAMがパンク。
1. チェックポイント：各ブロックの入口だけ覚え、内部の中間値は捨てる。
1. 逆伝播時：そのブロックだけ順伝播をやり直して中間値を復元し、正確な勾配を計算する。

## 学習の工学3：FlashAttention

Attentionは、文章の長さ $T$ に対して $T \times T$ の点数表を作ります。長さ8000なら6400万マス。これをGPUのメモリに書いて、また読んで……をくり返すのが遅さの原因でした。

**FlashAttention** は、点数表を**小さなタイルに分けて**、GPUの中の超高速な小さいメモリの中だけで計算を済ませてしまう方法です。結果は同じなのに、速く、メモリも少なく済みます。PyTorchでは `F.scaled_dot_product_attention` を使うと、使える環境なら自動でFlashAttention（などの高速な方法）が選ばれます。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
torch.manual_seed(0)
q, k, v = (torch.randn(2, 4, 128, 32) for _ in range(3))     # [B, ヘッド, T, hs]

att = q @ k.transpose(-2, -1) / math.sqrt(32)
att = att.masked_fill(torch.tril(torch.ones(128, 128)) == 0, float("-inf"))
manual = F.softmax(att, dim=-1) @ v

fast = F.scaled_dot_product_attention(q, k, v, is_causal=True)
print("手作りと同じ？", torch.allclose(manual, fast, atol=1e-5))
````

第15章の `CausalSelfAttention` の中の4行（点数→マスク→softmax→Vを掛ける）を、この1行に置き換えるだけで、GPUでは何倍も速くなります。自分のGPTを速くしたくなったら、最初に試すべき改造です。

<details><summary><b>基礎の補足：GPUのメモリ階層（SRAM vs HBM）：FlashAttentionが「近似なし」で速く省メモリな理由</b>（クリックで開く）</summary>

GPUの内部には、性格のまったく違う2種類のメモリがあります。

| メモリ | 例え | 容量の目安 | 読み書きの速さ |
|---|---|---|---|
| **HBM**（メインVRAM） | 広い書庫 | 数十GB（T4は約15GB） | 比較的遅い（ここへの往復がボトルネック！） |
| **SRAM**（チップ内キャッシュ） | 机の上 | 数十MB（ごく小さい） | **HBMの10倍以上 超高速！** |

- **標準Attention**：巨大な $T \times T$ 行列（$O(T^2)$）を丸ごとHBMに書き出し → マスクのためにまたHBMから読んで書き出し → softmaxのためにまた読み書き……と、遅い書庫を何度も往復します。
- **FlashAttention**：$Q, K, V$ を小さなタイルに分けて机の上（SRAM）へ運び、オンラインsoftmaxという工夫で**巨大な $T \times T$ 行列を一度もHBMに書き出さずに**正確な出力だけを計算します。その結果、HBMの消費は $O(T^2) \to O(T)$ に激減し、速度も数倍に跳ね上がります。

</details>

<details><summary><b>もっと深く：torch.compile</b>（クリックで開く）</summary>

`model = torch.compile(model)` の1行で、PyTorchがモデルの計算をまとめて最適化してくれます（初回の実行は準備に時間がかかります）。GPU上では、学習が数割速くなることがよくあります。

</details>

#### 📝 確認問題（`ch16-q13` · Lv.3 · FlashAttentionが速く省メモリな理由（HBM vs SRAMのメモリ階層））

**FlashAttention** が、通常のAttentionと**数学的に同じ結果**を出しながらずっと速く・省メモリになる本質的な理由は何ですか？

- **a.** 計算の重い指数関数やsoftmaxをすべて廃止し、単純な線形加算でAttention重みを代用するから
- **b.** $T \times T$ 行列を遅いHBMへ書き出さず、小タイルに分けて高速なSRAM上で計算を完結させるから
- **c.** 長文のAttention計算だけをネットワーク経由で別サーバーのCPU群へ自動分散して並列処理するから
- **d.** 過去トークンの約9割をランダムに間引いて無視し、総演算回数自体を10分の1へ削る近似手法だから

<details><summary>ヒント / 解答と解説を見る</summary>

- **ヒント 1**：GPUのボトルネックは「計算ユニットの速さ」よりも、メインメモリ（HBM）とのデータ往復にあることを思い出しましょう。
- **ヒント 2**：標準Attentionは $T \times T$ 行列をHBMへ何度も読み書きします。FlashAttentionはチップ内の超高速メモリ（SRAM）をどう活用するでしょうか？
- **正解**：`b`

**考え方**：標準Attentionは $Q K^\top$・マスク・softmax のたびに巨大な $T \times T$ 行列をHBMへ読み書きしますが、FlashAttentionはタイルごとにSRAM内で一気に処理して出力だけをHBMへ書き戻すため、メモリ使用量も $O(T^2)$ から $O(T)$ へ激減します。

1. GPUには「大容量だが遅い **HBM**（VRAM）」と「極小（数十MB）だが超高速な **SRAM**（チップ内キャッシュ）」がある。
1. 標準Attentionは $T \times T$ 行列をHBMに丸ごと作って何度も読み書きするため遅くメモリを食う。
1. FlashAttentionは $Q, K, V$ を小さなブロック（タイル）に分けてSRAMへ読み込み、オンラインsoftmaxで $T \times T$ をHBMに作らずに同じ答えを求める。

## 何百枚ものGPUで学習する：分散学習

1枚のGPUに乗らない、あるいは1枚では何年もかかる学習は、たくさんのGPUで分担します。この教科書の範囲（無料の環境）ではめったに使いませんが、論文やニュースを読むために、考え方だけは押さえておきましょう。

> **💡 理論のポイント：分散学習の4つの作戦**
>
> - **データ並列**（DDP）：GPUごとにモデルを**丸ごとコピー**し、違うデータで学習。勾配だけを全GPUで平均する。一番シンプルで、よく使われる。でも、モデルが1枚に乗らないと使えない
> - **ZeRO / FSDP**：データ並列の発展版。パラメータ・勾配・Adamの記録（第15章の16バイトの内訳）を、GPUたちで**分割して持ち合う**。必要なときだけ集めてくる。1枚に乗らない大きさのモデルも学習できる
> - **テンソル並列**：1つの行列の掛け算を、**行列を分割して**複数のGPUで計算する。1つの層が大きすぎるときに使う。GPU間の通信が多いので、同じマシン内のGPUどうしで使う
> - **パイプライン並列**：層ごとに担当のGPUを決める（1〜10層はGPU1、11〜20層はGPU2…）。工場の流れ作業。待ち時間（バブル）を減らす工夫が必要
>
> 最先端のLLMは、これらを全部組み合わせて（「3D並列」などと呼ばれます）、数千〜数万枚のGPUで学習されています。

> **¥0 無料枠のコツ：0円で分散学習を体験するなら**
>
> Kaggleのノートブックでは、無料で「T4×2」（GPU2枚）を選べることがあります。PyTorchの `torchrun` と DDP を使えば、2枚でのデータ並列を体験できます。興味がある人は、PyTorch公式の「DDPチュートリアル」を、Kaggleで試してみてください（[付録A](https://sakunokyumei.github.io/llm-research-textbook/a-setup.html)にKaggleの使い方があります）。

> **⚠️ つまずき注意：よくあるつまずき**
>
> - **KVキャッシュ付きの生成で結果が変わる**：位置の埋め込みの番号（`start_pos`）がずれていないか確認しましょう。キャッシュを使うときは、新しいトークンの位置は「これまでの長さ」から始まります。
> - **勾配累積で学習がおかしい**：損失を累積回数で割っていないと、学習率が実質何倍にもなってしまいます。また、`zero_grad()` は「更新した後」に1回だけ呼びます。
> - **`scaled_dot_product_attention` が遅い・使えない**：古いGPUや古いPyTorchでは、FlashAttentionが選ばれないことがあります。それでも結果は正しいので、心配はいりません。

#### 📝 確認問題（`ch16-q14` · Lv.3 · 分散学習の4方式（DDP・ZeRO/FSDP・テンソル並列・パイプライン並列））

- **a.** **KVキャッシュ**（過去のKey・Valueテンソル再利用）
- **b.** **ZeRO / FSDP**（Fully Sharded Data Parallel）
- **c.** **top-p**（Nucleus Sampling：累積確率サンプリング）
- **d.** 単純な**データ並列**（DDP: Distributed Data Parallel）

<details><summary>ヒント / 解答と解説を見る</summary>

1. DDP：各GPUにモデル丸ごとコピー＋違うデータを担当。
1. ZeRO / FSDP：重み・勾配・Adam状態をGPU間で分割保持（シャード）し、必要時に通信で集める。
1. テンソル並列：1つの行列積自体を複数GPUに分割。
1. パイプライン並列：1〜10層をGPU1、11〜20層をGPU2のように層ごとに分担。
