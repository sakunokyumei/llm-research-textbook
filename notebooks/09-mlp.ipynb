````python
import re, math, urllib.request
import torch
import torch.nn as nn
import torch.nn.functional as F

def load_soseki(file_id="773_ruby_5968"):
    url = ("https://raw.githubusercontent.com/aozorahack/aozorabunko_text/0984f7dc18a2270bdfd2ff99bf14fd1b3ebb4f65/"
           f"cards/000148/files/{file_id}/{file_id}.txt")
    try:
        with urllib.request.urlopen(url, timeout=30) as res:
            text = res.read().decode("cp932")
    except Exception as e:
        raise RuntimeError(f"青空文庫の取得に失敗しました（時間をおいて再実行するか付録Eを参照）: {e}") from e
    text = text.split("-------------------------------------------------------")[2]
    text = text.split("底本：")[0]
    text = re.sub(r"《.*?》", "", text)
    text = re.sub(r"※?［＃.*?］", "", text)
    text = text.replace("｜", "").replace("\r\n", "\n")
    text = re.sub(r"\n{3,}", "\n\n", text)
    return text.strip()

text = load_soseki()
chars = sorted(set(text))
V = len(chars)
stoi = {ch: i for i, ch in enumerate(chars)}
itos = {i: ch for ch, i in stoi.items()}
data = torch.tensor([stoi[c] for c in text])
n = int(len(data) * 0.9)
train_data, val_data = data[:n], data[n:]
````

## 8文字の窓

文章の上に「8文字分の窓」を置いて、窓の中の8文字を入力、その次の1文字を正解にします。窓を1文字ずつずらせば、問題がたくさん作れます。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
T = 8   # 窓の大きさ（何文字前まで見るか）

def get_batch(data, batch_size, g):
    ix = torch.randint(0, len(data) - T, (batch_size,), generator=g)  # 窓の開始位置をランダムに
    x = torch.stack([data[i:i + T] for i in ix])     # 入力：8文字（stack は、同じ形のテンソルを積み重ねて1つにする）
    y = data[ix + T]                                 # 正解：その次の1文字
    return x, y

g = torch.Generator().manual_seed(0)
x, y = get_batch(train_data, 4, g)
print(x.shape, y.shape)
for i in range(4):
    print(repr("".join(itos[j] for j in x[i].tolist())), "→", repr(itos[y[i].item()]))
````

## 埋め込み：文字を「点」にする

ニューラルネットワークに文字を渡すには、文字を数にしなければなりません。第6章のone-hot（2061個の数のうち1つだけ1）は、無駄が多いうえに、どの文字どうしも「同じくらい違う」ものとして扱ってしまいます。

そこで、各文字に **少数の数の組** を割り当てます。たとえば2個の数なら、各文字は平面上の1つの点になります。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
torch.manual_seed(0)
C = torch.randn(V, 2)          # 2061文字 × 2個の数（最初はでたらめ）
print("「先」の座標：", C[stoi["先"]])
print("「奥」の座標：", C[stoi["奥"]])
print(C[x].shape)              # 4問 × 8文字 × 2個の数
````

> **💡 理論のポイント：埋め込み（embedding）**
>
> 文字（やトークン）ごとに割り当てた数の組（ベクトル）を **埋め込み**（embedding）と言います。表 `C` の各行が、その文字の埋め込みです。
>
> ポイントは、**この座標も学習で決める**こと。最初はでたらめな位置ですが、学習が進むと「同じような場面で使われる文字」が近くに集まってくるように動いていきます。「先」と「奥」が近くに来れば、「先生と」を見て学んだことが「奥さんと」にも少し役立つようになる。数えるモデルにはできなかった **汎化**（はんか、generalization：見たことのないものへの応用）が起きるのです。
>
> 実際のモデルでは、2個ではなく数十〜数千個の数を使います（この章では32個、Qwen2.5-0.5Bでは896個）。たくさんの数を使えば、「品詞っぽさ」「漢字っぽさ」「人っぽさ」……のように、たくさんの特徴を同時に表せます。

`C[x]` の形 `[4, 8, 2]` は、「4問 × 8文字 × 各文字2個の数」です。第6章で、one-hot × 表 ＝ 表の行の取り出し、でしたね。`C[x]` は、それを番号で直接やっています。

<details><summary><b>基礎の補足：埋め込みルックアップ `C[x]` vs one-hot 行列積</b>（クリックで開く）</summary>

| 比較項目 | `F.one_hot(x, V).float() @ C` | `C[x]`（＝ `nn.Embedding(V, d)(x)`） |
|---|---|---|
| 計算結果 | `C` の `x` 行目を取り出す | **数学的に完全に同じ！** |
| 計算量・メモリ | 2060個の `0` にも掛け算して巨大行列を作る | 行を直接引くだけ（圧倒的に高速・省メモリ） |
| 形の変化 | `[B, T]` → `[B, T, V]` → `[B, T, d]` | `[B, T]` → `[B, T, d]` |

`x` の1マス（文字ID）が、そのまま長さ `d` のベクトル（座標）に置き換わるので、形は常に `[B, T] → [B, T, d]` になります。

</details>

#### 📝 確認問題（`ch09-q02` · Lv.2 · 埋め込み C[x] と one-hot × C の関係、および汎化が生まれる理由）

埋め込み表 `C`（形 `[V, d]`）に対して `C[x]`（または `nn.Embedding(V, d)(x)`）を行うことと、one-hot 表現を直接使うことの違いとして**正しい**のはどれですか？

- **a.** 埋め込み表現を使っても文字 ID が隣り合う文字（例：ID 10 と ID 11）同士しか近くになれず、ID が離れた似た役割の文字は表現できない

- **c.** `C[x]` で得られる密な埋め込みベクトルは one-hot ベクトルより次元数が必ず大きくなり、語彙数 $V$ 以上の膨大なメモリを消費する
- **d.** `C[x]` は `one_hot(x) @ C` と同じ計算を高速に行い、各文字を学習可能な密な座標へ写して「似た文字が近くに集まる＝汎化」を可能にする

- **ヒント 1**：one-hot 表現ではすべての文字ペアが直交して無関係扱いになりますが、埋め込み空間では「似た文脈に出る文字」がどう配置されるでしょう？
- **ヒント 2**：`C[x]` は `one_hot(x) @ C` と数学的に同一の行取り出しを高速に行い、学習で似た文字を近い座標へ集めることで未知の並びにも汎化できます。

1. 計算面：`C[x]` ＝ `one_hot(x) @ C`（無駄な0の掛け算を省略した行ルックアップ）。
1. 表現面：2061次元のスカスカな one-hot を、特徴が詰まった 32次元の座標へ圧縮。
1. 汎化面：似た役割の文字が近い座標に集まるため、未知の8文字並びにも推測が効く。

## 層を重ねる：多層パーセプトロン

8文字ぶんの埋め込みが手に入ったら、それを1列に並べて、計算を2段重ねます。

1. **隠れ層**：埋め込みを並べたもの（8文字 × 32 = 256個の数）を、256個の「ニューロン」で変換する
2. **出力層**：その結果から、2061文字それぞれの点数（ロジット）を出す

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
d, H = 32, 256    # 埋め込みの大きさ、隠れ層の大きさ
torch.manual_seed(0)
C = torch.randn(V, d)
W1 = torch.randn(T * d, H) * 0.1
b1 = torch.zeros(H)
W2 = torch.randn(H, V) * 0.1
b2 = torch.zeros(V)

emb = C[x]                       # [4, 8, 32]
flat = emb.view(4, T * d)        # [4, 256]：8文字ぶんを横一列に並べる
h = torch.tanh(flat @ W1 + b1)   # [4, 256]：隠れ層
logits = h @ W2 + b2             # [4, 2061]：次の文字の点数
print(emb.shape, flat.shape, h.shape, logits.shape)
````

```text
torch.Size([4, 8, 32]) torch.Size([4, 256]) torch.Size([4, 256]) torch.Size([4, 2061])
```

</details>

> **🐍 Python・PyTorchのポイント：.view と .reshape で形を変える**
>
> `emb.view(4, T * d)` は、中身の数はそのままで、形だけを `[4, 8, 32]` から `[4, 256]` に変えます。8文字ぶんの32個ずつを、横一列（256個）に並べ直すイメージです。`-1` を使うと「残りは自動で計算して」という意味になり、`emb.view(4, -1)` でも同じ結果になります。なお、`transpose` 後などでメモリが不連続なテンソルでは `.view()` がエラーになるため、安全な `.reshape(4, -1)`（第6章・[付録B](https://sakunokyumei.github.io/llm-research-textbook/b-python.html)）を使うこともよくあります。

<details><summary><b>基礎の補足：`.view(B, -1)` の形の算数と、MLP全体の shape パイプライン</b>（クリックで開く）</summary>

`.view()` は**要素の総数を保ったまま**形だけを並べ替える操作です。`emb` の形が `[B, T, d] = [4, 8, 32]` のとき：
- 総要素数は $4 \times 8 \times 32 = 1024$ 個
- `emb.view(4, -1)` は「第0軸を `4` に固定し、残りの第1軸は自動で割ってね」の意味 → $1024 \div 4 = 256$（ちょうど $T \times d = 8 \times 32$）！
- コードで `e.view(len(x), -1)` と書けば、バッチサイズ `B` が訓練時(`256`)・検証時(`16239`)・生成時(`1`)で変わっても同じ1行で動きます。

MLP全体のテンソル形の流れを1本の地図で押さえておきましょう：

| ステップ | 計算 | テンソルの形（`shape`） | 意味 |
|---|---|---|---|
| 入力 | `x` | `[B, T]` = `[4, 8]` | $B$ 問 × 直前 $T$ 文字のID |
| 埋め込み | `e = C[x]` | `[B, T, d]` = `[4, 8, 32]` | 各文字を $d$ 次元の座標に変換 |
| 平坦化 | `flat = e.view(B, -1)` | `[B, T*d]` = `[4, 256]` | 8文字ぶんを横1列に連結 |
| 隠れ層 | `h = tanh(flat @ W1 + b1)` | `[B, H]` = `[4, 256]` | $H$ 個のニューロンで特徴抽出 |
| 出力層 | `logits = h @ W2 + b2` | `[B, V]` = `[4, 2061]` | 全語彙 $V$ 文字の生の点数 |

</details>

> **💡 理論のポイント：ニューロンと全結合層**
>
> `flat @ W1 + b1` が、ニューラルネットワークの基本部品 **全結合層**（linear layer）です。数式では
>
> $$\boldsymbol{y} = \boldsymbol{x} W + \boldsymbol{b}$$
>
> と書きます。出力の1つ1つ（ニューロン）は、「入力の数それぞれに重みを掛けて全部足し、最後に $b$（バイアス）を足したもの」です。
>
> <div class="math-display">ニューロン ＝ 入力₁×重み₁ ＋ 入力₂×重み₂ ＋ … ＋ 入力₂₅₆×重み₂₅₆ ＋ バイアス</div>
>
> 1つのニューロンは「どの特徴をどれくらい重視するか」の投票係のようなものです。256個のニューロンが、それぞれ違う観点で8文字を眺めて、点数をつけます。行列の掛け算 `@` は、この256個のニューロンの計算を一度にやってしまう技でした（第6章）。
>
> この「全結合層を何段か重ねたもの」を **多層パーセプトロン**（multilayer perceptron、**MLP**）と呼びます。

#### 📝 確認問題（`ch09-q04` · Lv.2 · .view(B, -1) の形変換と要素数保存の計算）

テンソル `emb` の形が `[8, 6, 16]`（8問 × 6文字 × 16次元）のとき、`flat = emb.view(8, -1)` を実行すると `flat` の形は `[8, ?]` になります。`?` に入る整数はいくつですか？

#### 📐 数学のポイント：なぜ非線形（`tanh`）が必須か：2×2の具体数で「線形崩壊」を証明する

活性化関数がないと、全結合層を何段重ねても、結局は**1段の全結合層**に潰れてしまいます。

$$(xW_1)W_2 = x(W_1 W_2) = xW'$$

**具体的な2×2の数字で確かめてみましょう！** 入力 $x = [1, 2]$、第1層 $W_1 = \begin{pmatrix} 1 & 2 \\ 0 & 1 \end{pmatrix}$、第2層 $W_2 = \begin{pmatrix} 2 & -1 \\ 1 & 3 \end{pmatrix}$ とします。

- **活性化なし（2段を順番に計算）**：
  1. 第1層：$x W_1 = [1\cdot 1 + 2\cdot 0, \; 1\cdot 2 + 2\cdot 1] = [1, 4]$
  2. 第2層：$(x W_1) W_2 = [1\cdot 2 + 4\cdot 1, \; 1(-1) + 4\cdot 3] = [6, 11]$
- **1つの行列 $W' = W_1 W_2$ に合体して1段で計算**：
  - $W' = \begin{pmatrix} 1 & 2 \\ 0 & 1 \end{pmatrix}\begin{pmatrix} 2 & -1 \\ 1 & 3 \end{pmatrix} = \begin{pmatrix} 4 & 5 \\ 1 & 3 \end{pmatrix}$
  - $x W' = [1\cdot 4 + 2\cdot 1, \; 1\cdot 5 + 2\cdot 3] = [6, 11]$（**どんな $x$ を入れても完全に一致！**）

```python no-run
import math
# 活性化なしでは (x @ W1) @ W2 == x @ (W1 @ W2) で1層に潰れる
# 一方 tanh を間に挟むと…
h = [math.tanh(1), math.tanh(4)]   # [0.7616, 0.9993]（まっすぐの比率が曲がる！）
y = [h[0]*2 + h[1]*1, h[0]*(-1) + h[1]*3]  # [2.5225, 2.2364] → どんな1枚の行列 W' でも再現不可！
```

たとえば「数を2倍して、さらに3倍する」操作は、最初から「6倍する（$2 \times 3 = 6$）」操作1回と同じですよね。掛け算だけなら、何十層重ねようが結局1段に潰れてしまいます。

そこで、層と層の間に「**直線ではない**（非線形な）」関数を挟みます。折れ曲がったり、S字になったりする関数を途中に挟むことで、1回にまとめることができなくなり、複雑で曲がりくねった現実の言葉のパターンを自在に表現できるようになります。

- **tanh**（ハイパボリックタンジェント）：-1から1の間のS字。この章で使う、昔ながらの関数
- **ReLU**（レルー）：マイナスは0、プラスはそのまま。単純で速いので大人気
- **GELU**（ジェルー）：ReLUをなめらかにしたもの。GPTで使われている（第12章）

- **b.** $(x W_1) W_2 = x (W_1 W_2) = x W'$ となり、何層重ねても結局 1 段の全結合層（線形変換）と同じ表現力に潰れてしまう

def vec_mat(x, W):

    h = [math.tanh(v) for v in vec_mat(x, W1)]
    return vec_mat(h, W2)

## 部品をまとめる：クラスで書く

ここまでの部品（`C`、`W1`、`b1`、`W2`、`b2`）をバラバラに持っていると、管理が大変です。PyTorchでは、モデルを **クラス** という設計図にまとめて書きます。

PyTorchの `nn.Module` に入る前に、まずは**ふつうのPythonだけの最小コード**で `class`・`__init__`・`self`・継承・`super().__init__()` をスッキリ理解しておきましょう！

<details><summary><b>基礎の補足：理解のための最小コード：PyTorchの前に「ふつうのPythonクラス」を押さえる</b>（クリックで開く）</summary>

**① 基本のクラス（設計図）と `__init__`・`self`**：
関数を「部品（データ）と処理のセット」にまとめたものが **クラス** です。

```python no-run
class LinearRule:                  # 「LinearRule」という名前の設計図
    def __init__(self, a, b):      # 実物を作るときに1回だけ呼ばれる初期化関数
        self.a = a                 # self（＝作られた実物自身）に部品 a を保存
        self.b = b                 # 部品 b を保存

    def calc(self, x):             # 実物が持つメソッド（関数）
        return self.a * x + self.b # 保存しておいた self.a, self.b を使って計算

rule = LinearRule(3, 1)            # 設計図から実物（インスタンス）を1個作る
print(rule.calc(2))                # → 3 * 2 + 1 = 7
```

**② 継承 `class 子(親):` と `super().__init__()`**：
すでにある便利な設計図（親クラス）の機能を丸ごと引き継いで、自分用の部品を足す仕組みを **継承** と呼びます。

```python no-run
class BaseTool:
    def __init__(self):
        self.registry = []         # 親が用意する管理台帳

class MyTool(BaseTool):            # BaseTool の機能をそのまま引き継ぐ！
    def __init__(self, scale):
        super().__init__()         # 親（BaseTool）の __init__ を必ず先に実行して台帳を作る
        self.scale = scale

    def forward(self, x):
        return x * self.scale
```

**③ PyTorch の `class MLP(nn.Module):` との1対1対応**：
- `class MLP(nn.Module):` → PyTorchの親クラス `nn.Module` を継承（すると `model.parameters()` や `.to("cuda")` や `state_dict()` が全部自動で使えるようになる！）
- `super().__init__()` → 親 `nn.Module` のパラメータ管理台帳を初期化（忘れるとエラー！）
- `self.emb = ...` → 必ず `self.` を付けて代入することで、台帳に部品が自動登録される
- `def forward(self, x):` → 順伝播の計算。使うときは `model.forward(x)` ではなく **`model(x)`** と呼ぶのが作法

</details>

それでは、MLPを `nn.Module` のクラスにまとめてみましょう。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
class MLP(nn.Module):
    def __init__(self, V, T, d=32, H=256):
        super().__init__()
        self.emb = nn.Embedding(V, d)     # 埋め込みの表（C）
        self.fc1 = nn.Linear(T * d, H)    # 隠れ層（W1, b1）
        self.fc2 = nn.Linear(H, V)        # 出力層（W2, b2）

    def forward(self, x):
        e = self.emb(x)                          # [B, T, d]
        h = torch.tanh(self.fc1(e.view(len(x), -1)))  # [B, H]
        return self.fc2(h)                       # [B, V]

torch.manual_seed(0)
model = MLP(V, T)
print(model)
print("パラメータの数：", sum(p.numel() for p in model.parameters()))
````

## 学習させる

学習ループは第8章とほぼ同じです。今回は、パラメータの更新を **オプティマイザ**（optimizer）という道具に任せます。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
@torch.no_grad()
def val_loss(model):
    X = torch.stack([val_data[i:i + T] for i in range(len(val_data) - T)])
    Y = val_data[T:]
    return F.cross_entropy(model(X), Y).item()

torch.manual_seed(0)
g = torch.Generator().manual_seed(0)
model = MLP(V, T)
optimizer = torch.optim.SGD(model.parameters(), lr=0.1)

for step in range(10001):
    x, y = get_batch(train_data, 256, g)
    loss = F.cross_entropy(model(x), y)
    optimizer.zero_grad()     # 勾配を0に
    loss.backward()           # 勾配を計算
    optimizer.step()          # パラメータを更新
    if step % 2000 == 0:
        print(f"ステップ {step:5d}：訓練 {loss.item():.3f}  検証 {val_loss(model):.3f}")
````

> **🐍 Python・PyTorchのポイント：オプティマイザ**
>
> `optimizer.zero_grad()` → `loss.backward()` → `optimizer.step()` の3行は、第7章で手で書いた「勾配を0に → 勾配を計算 → `w -= lr * w.grad`」と同じことをしています。`torch.optim.SGD` は、モデルの全パラメータについて、まとめて `w -= lr * w.grad` をやってくれる係です。<br>
> `@torch.no_grad()` を関数の上に付けると、その関数の中身全体が `with torch.no_grad():` の中にあるのと同じになります（関数の上に置く `@` は **デコレータ** という印で、第6章の行列の掛け算の `@` とは別物です）。

#### 📝 確認問題（`ch09-q12` · Lv.2 · torch.optim.SGD の3行と @torch.no_grad() デコレータの意味）

`optimizer = torch.optim.SGD(model.parameters(), lr=0.1)` を使った3行 `optimizer.zero_grad()` → `loss.backward()` → `optimizer.step()` は、第7章の手書きループの何に対応していますか？

- **a.** `zero_grad()`＝全パラメータの勾配リセット、`backward()`＝勾配計算、`step()`＝`with torch.no_grad():` 内で全パラメータを `p -= lr * p.grad` で更新
- **b.** `zero_grad()`＝全パラメータの重みを 0 に初期化、`backward()`＝検証データの損失計算、`step()`＝学習済みモデルで文章を 1 文字だけ生成
- **c.** `zero_grad()`＝勾配のリセット、`step()`＝内部で自動的に `loss.backward()` と重み更新の両方を実行するため、実は `loss.backward()` の行は削除してよい
- **d.** `zero_grad()`＝ミニバッチの抽出、`backward()`＝誤差逆伝播を使わずランダム探索で候補生成、`step()`＝一番損失の低い重みを採用して上書き

<details><summary>ヒント / 解答と解説を見る</summary>

- **ヒント 1**：第7〜8章ではパラメータが1個だったので手で勾配リセットと `W -= lr * W.grad` を書けましたが、MLP では部品が5個以上に増えました。
- **ヒント 2**：オプティマイザは新しい計算ではなく、`zero_grad()` で全パラメータの `.grad` を消し、`step()` で `no_grad` 内の更新をまとめて代行しています。
- **正解**：`a`

**考え方**：オプティマイザは新しい魔法ではなく、第7章で手で書いた「勾配リセット」と「`no_grad` 内での更新」を全パラメータぶん自動化してくれる道具です。

1. `optimizer.zero_grad()`：登録された全パラメータの `.grad` をリセット。
1. `loss.backward()`：計算グラフを逆にたどり全パラメータの `.grad` を計算。
1. `optimizer.step()`：全パラメータに `p -= lr * p.grad` を実行。

## 書かせてみる

````python
@torch.no_grad()
def generate(model, start, n=100, seed=0):
    g = torch.Generator().manual_seed(seed)
    ids = [stoi[c] for c in start]
    for _ in range(n):
        x = torch.tensor([ids[-T:]])               # 直前8文字
        probs = F.softmax(model(x), dim=1)[0]
        ids.append(torch.multinomial(probs, 1, generator=g).item())
    return "".join(itos[i] for i in ids)

print(generate(model, "私はその人を常に"))
````
