## 掃除する：正規表現

機械学習では、**データの掃除（前処理）がとても大事**です。「ゴミを入れたらゴミが出てくる（Garbage In, Garbage Out）」という格言があるくらいです。ルビを消すには **正規表現**（regular expression）を使います。正規表現は「こういう形の文字列」というパターンを書くための小さな言語です。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
import re

sample = "私《わたくし》はその人を常に先生《せんせい》と呼んでいた。"
print(re.sub(r"《.*?》", "", sample))
````

#### 🐍 Python・PyTorchのポイント：正規表現の読み方

`re.sub(パターン, 置き換える文字, 対象)` は「パターンに合う部分を全部置き換える」関数です。`r"《.*?》"` は次のように読みます。

| 部分 | 意味 |
|---|---|
| `《` | 「《」という文字 |
| `.` | 何でもいい1文字 |
| `*` | 直前のものが0回以上くり返す |
| `?` | ただし、なるべく短く（控えめに） |
| `》` | 「》」という文字 |

つまり「《 から始まって、いちばん近い 》 で終わる部分」です。`?` を付け忘れると「なるべく長く」になって、最初の《から**最後の**》までをごっそり消してしまいます。試してみましょう。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
print(re.sub(r"《.*》", "", sample))
````

| パターン | マッチする区間（中間） | 置換後の出力（`str`型） | 理由 |
|---|---|---|---|
| `r"《.*?》"`（最短一致） | ①`《わたくし》`、②`《せんせい》` の2か所 | `"私は先生だ"` | `?` があるので一番近い `》` で止まる |
| `r"《.*》"`（最長一致） | `《わたくし》は先生《せんせい》` の1か所 | `"私だ"` | `?` がないので最後の `》` まで欲張って飲み込む |

では、本文全体を掃除する関数を作ります。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
def clean_aozora(text):
    text = text.split("-------------------------------------------------------")[2]  # 冒頭の説明を除く
    text = text.split("底本：")[0]                   # 末尾の書誌情報を除く
    text = re.sub(r"《.*?》", "", text)              # ルビを除く
    text = re.sub(r"※?［＃.*?］", "", text)          # 入力者注を除く
    text = text.replace("｜", "").replace("\r\n", "\n")
    text = re.sub(r"\n{3,}", "\n\n", text)           # 3行以上の空行を2行に
    return text.strip()

text = clean_aozora(raw.decode("cp932"))
print(len(text), "文字")
print(text[:150])
````

非貪欲（最短一致）の `?` の働きを確認します。次のコードを実行すると何が表示されますか？

<details><summary>ヒント / 解答と解説を見る</summary>

- **ヒント 1**：正規表現の `.*?`（非貪欲・最短一致）は、最初の `《` から一番近い `》` でマッチを止めます。

**考え方**：`.*?` は最短一致（控えめなマッチ）なので、最初の `《` から最も近い `》` までを1つのまとまりとして置き換えます。そのため間の「とB」は残ります。

1. 1回目のマッチ：`《あ》` → 空文字 `""` に置換。
1. 間の `とB` はパターン外なのでそのまま残る。
1. 2回目のマッチ：`《びー》` → 空文字 `""` に置換。
1. 結果：`AとB`。

#### 📐 数学のポイント：条件付き確率 $P(B \mid A)$：9ステップで理由から理解する

**① 解決したい具体的な問題**：「今の文字が『先』だ」と分かっているとき、次の文字が『生』『へ』『刻』…である確率を正しく出したい。

**② 素朴な方法（単なる出現回数・全体割合）ではなぜダメなのか？**
- もし文章全体の回数ランキング（**unigram**）だけを使うと、『こころ』で最多は「の」（7556回）なので、どんな文字の次でも「の」が一番選ばれやすくなり、「ののたのい…」のような日本語にならない文が出てしまいます。
- また、生の回数 `597` のままでは「678回中597回（88%）」なのか「1万回中597回（6%）」なのか分からず、合計1の確率分布にもなりません。

**③ 新しい発想**：小説全体を見るのではなく、**「1文字目が $a$ であるペアだけ」に世界を絞り込み**、その中で次が $b$ だった割合をとる。これが **条件付き確率** です。

**④ 記号の読み方表**：
| 記号 | 読み方 | 意味 |
|---|---|---|
| $P(B \mid A)$ | 「AのもとでのBの確率」 | 条件 $A$ が起きた世界で、出来事 $B$ が起きる割合 |
| $\mid$（縦棒） | 「〜という条件のもとで」 | **右側**が分かっている条件（分母）、**左側**が予測したいもの |
| $\text{count}(a, b)$ | 「$a$ の次に $b$ が来た回数」 | 隣接ペア `(a, b)` の出現回数 |
| $\text{count}(a)$ | 「$a$（の後に文字がある箇所）の回数」 | 条件 $a$ で始まるペアの総数 $\sum_b \text{count}(a, b)$ |

**⑤ 小さな10文字コーパスで手計算**：`s = "あいうあいうあいあえ"`（10文字・隣接9ペア）で、今の文字が「あ」のときを手計算してみましょう。
- 全9ペア：`あい, いう, うあ, あい, いう, うあ, あい, いあ, あえ`
- 「あ」で始まるペアだけ絞り込む：`あい`（3回）、`あえ`（1回）→ 合計4回
- したがって：
$$P(\text{い} \mid \text{あ}) = \frac{3}{4} = 0.75, \qquad P(\text{え} \mid \text{あ}) = \frac{1}{4} = 0.25$$
- 合計は $0.75 + 0.25 = 1.00$。きちんと確率分布の2条件（0以上・合計1）を満たします！

**⑥ 一般式**：『こころ』のような長い文章でも同じです。
$$P(b \mid a) = \frac{\text{count}(a, b)}{\text{count}(a)} = \frac{\text{count}(a, b)}{\sum_{b'} \text{count}(a, b')}$$
『こころ』の「先」なら：
$$P(\text{生} \mid \text{先}) = \frac{\text{「先生」の回数}}{\text{「先」の回数}} = \frac{597}{678} \approx 0.88$$

**⑦ Pythonの最小コードで検算**：
```python no-run
from collections import Counter
s = "あいうあいうあいあえ"
p = Counter(zip(s, s[1:]))
a_total = sum(n for (a, b), n in p.items() if a == "あ")   # 4
print("P(い|あ) =", p[("あ", "い")] / a_total)             # 0.75
print("P(え|あ) =", p[("あ", "え")] / a_total)             # 0.25
```

**⑧ よくある誤解**：$P(B \mid A)$ と $P(A \mid B)$ は**まったく別物**です。『こころ』では $P(\text{生} \mid \text{先}) = 597/678 \approx 0.88$ ですが、逆向きの $P(\text{先} \mid \text{生})$ は「生」の次に「先」が来る確率（たった4回）なのでほぼ $0$ です。縦棒 $\mid$ の**右側が条件（分母）**と意識しましょう。

**⑨ 言語モデルとの接続**：言語モデルとはまさに $P(\text{次の文字} \mid \text{それまでの文字})$ を出す仕組み（第3章）でした。今回は直前の**1文字だけ**を条件にして2つ（bi）の文字の組を数えるので、**bigram**（バイグラム）モデルと呼びます。

## 漱石に書かせる

数えた結果を使えば、文章が作れます。今の文字の「次の文字の回数」をそのまま重みにして、くじを引くのです。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
import random

following = {}                       # 「文字 → その次の文字の回数」の辞書
for a, b in zip(text, text[1:]):      # 隣り合う2文字 a, b を順番に
    if a not in following:            # a が初めて出てきたら
        following[a] = Counter()      #   a 用の空のカウンターを用意
    following[a][b] += 1              # 「a の次に b が来た」回数を1増やす

def generate(start, n=80):
    out = start
    ch = start[-1]                    # 書き出しの最後の1文字から続ける
    for _ in range(n):
        c = following[ch]
        ch = random.choices(list(c.keys()), weights=list(c.values()))[0]
        out += ch
    return out

random.seed(1)
print(generate("私"))
````

文字列 `text` と条件となる1文字 `target` を受け取り、`target` の次に来る各文字 `b` に対する条件付き確率 $P(b \mid \text{target})$ を辞書 `{b: 確率}` として返す関数 `bigram_probs(text, target)` を自力実装してください（`target` の次に続く文字が1つもない場合は空辞書 `{}` を返す）。

**ひな形コード：**
```python
from collections import Counter

def bigram_probs(text, target):
    # 1. zip(text, text[1:]) を回し、a == target のときの b を Counter で数える
    # 2. 合計 total = sum(counts.values()) を求める
    # 3. total == 0 なら {} を返す
    # 4. {b: n / total for b, n in counts.items()} を返す
    pass
```

<details><summary>ヒント / 解答と解説を見る</summary>

- **ヒント 1**：`zip(text, text[1:])` で隣接ペア `(a, b)` を取り出し、`a == target` のときだけ `b` の回数を `Counter` で数えます。
- **ヒント 2**：合計 `total = sum(counts.values())` が `0` なら空辞書 `{}` を返し、そうでなければ各文字を `n / total` にした辞書を返します。

**解答例：**
```python
from collections import Counter

def bigram_probs(text, target):
    counts = Counter()
    for a, b in zip(text, text[1:]):
        if a == target:
            counts[b] += 1
    total = sum(counts.values())
    if total == 0:
        return {}
    return {b: n / total for b, n in counts.items()}
```

**検算テスト（Colabで実行して確認）：**

```python
assert bigram_probs("あいう", "う") == {}
assert bigram_probs("あいう", "ん") == {}
print("ALL PASSED")
```

**考え方**：条件 `a == target` で絞り込んだ `Counter` を作り、その合計 `total` で割ることで、定義式 $P(b \mid a) = \text{count}(a, b) / \text{count}(a)$ をそのままコードで表現できます。

1. `counts = Counter()` を用意し、`for a, b in zip(text, text[1:]):` で `a == target` なら `counts[b] += 1`。
1. `total = sum(counts.values())` を計算。`0` ならゼロ除算を防ぐため `{}` を返す。
1. `{b: n / total for b, n in counts.items()}` で各確率を辞書にして返す。

## なぜ1文字では足りないのか

> **💡 理論のポイント：マルコフ性と n-gram**
>
> bigramモデルは「次の文字は、**直前の1文字だけ**で決まる」と仮定しています。この「直前だけを見る」性質を **マルコフ性** と言います（「マルコフ連鎖」という名前を聞いたことがあるかもしれません）。
>
> もっと前まで見れば、もっと自然になるはずです。直前2文字なら **trigram**（トライグラム）、一般に直前 $n-1$ 文字を見るモデルを **n-gram** モデルと呼びます。2010年代の半ばごろまで、機械翻訳や音声認識の中心はこのn-gramモデルでした。

では、直前2文字を見るtrigramを作ってみましょう。

🔎 **実行前に予想しよう**：このセルを実行すると、どんな数値・文字列・テンソルの形（shape）やグラフが出るか予想してから ▶ を押してください。

````python
following3 = {}
for a, b, c in zip(text, text[1:], text[2:]):
    key = a + b
    if key not in following3:
        following3[key] = Counter()
    following3[key][c] += 1

def generate3(start, n=80):
    out = start
    for _ in range(n):
        c = following3.get(out[-2:])  # 最後の2文字の「次の文字の回数」。辞書になければ None
        if c is None:                 # 続きが見つからなければ
            break                     #   ループを途中で抜けて終わり
        out += random.choices(list(c.keys()), weights=list(c.values()))[0]
    return out

random.seed(1)
print(generate3("私は"))
print("2文字の組み合わせの種類：", len(following3))
````

「先生を見詰めている」「Ｋに向って」「奥さんにはまた愛する」……かなり『こころ』です！ 直前2文字を見るだけで、ぐっと文章らしくなりました。じゃあ、10文字見れば完璧では？

> **💡 理論のポイント：データが足りない問題（スパース性）**
>
> そうはいきません。見る文字数を増やすと、**組み合わせの数が爆発**します。語彙が2061文字なら、組み合わせの数は文字数に対して指数関数的に膨れ上がります。
>
> - 直前1文字（bigram）：$2061^2 \approx 425$万通り（『こころ』に出てくるのは約2万種類だけ）
> - 直前2文字（trigram）：$2061^3 \approx 87.5$億通り
> - 直前10文字（11-gram）：$2061^{11} \approx 2.85 \times 10^{36}$通り（地球上の砂粒の数 $\sim 10^{19}$〜$10^{20}$ を桁違いに上回る天文学的数字！ ちなみに観測可能な宇宙全体の原子数は約 $10^{80}$ 個で、さらに25-gramあたりでそれも超えます）
>
> 『こころ』全体でもたった16万字しかありません。文字数を少し増やしただけで、組み合わせの大部分はデータの中に**一度も出てこなくなります**。
>
> 一度も見たことがない並びに出会うと、数えるだけのモデルは「確率0」と答えるしかありません（上の `generate3` が `None` で止まるのは、このときです）。表のマス目の大半が0でスカスカになってしまうこの現象を、**スパース性**（まばらさ）の問題と言います。
>
> 人間なら、「先生は」を見たことがあれば「教授は」の次も似たような言葉が来そうだと推測できます。「先生」と「教授」が**似ている**ことを知っているからです。数えるだけのモデルには、この「似ている」が分かりません。
>
> この壁を越えるために、文字を「数の組（ベクトル）」で表して、似た文字を近くに置く方法を学びます。それがニューラルネットワークの出番であり、第9章の「埋め込み」です。

#### 📝 確認問題（`ch04-q11` · Lv.2 · マルコフ性と bigram / trigram / n-gram の関係）

直前の **2文字** を見て次の1文字を予測するモデル（例：$P(c \mid a, b)$）は何と呼ばれ、一般の $n$-gram モデルは何文字を条件にしますか？

- **a.** 直前2文字を見るのは **unigram（1-gram）** モデル。一般の $n$-gram は文全体に含まれる単語の総数を数えて予測する
- **b.** 直前2文字を見るモデルは有限の過去しか見ないためマルコフ性を持たず、$n$-gram モデルの仲間には含まれない
- **c.** 直前2文字を見るのは **trigram（3-gram）**。一般の $n$-gram は直前 **$n - 1$ 文字** を条件にする
- **d.** 直前2文字を見るのは **bigram（2-gram）** モデル。一般の $n$-gram は直前 **$n$ 文字** を条件にして次の1文字を予測する

<details><summary>ヒント / 解答と解説を見る</summary>

- **ヒント 1**：$n$-gram の $n$ は「条件の文字数」ではなく「条件＋予測する次の1文字」の合計文字数です。
- **ヒント 2**：「条件2文字＋次の1文字＝合計3文字の組」なので 3-gram（trigram）、一般に $n$-gram の条件は $n-1$ 文字です。
- **正解**：`c`

**考え方**：bigram は `(a, b)` の2文字組（条件1文字）、trigram は `(a, b, c)` の3文字組（条件2文字）、一般に $n$-gram は直前 $n-1$ 文字を条件にします。

1. unigram ($n=1$)：条件0文字、$P(a)$。
1. bigram ($n=2$)：条件1文字、$P(b \mid a)$。
1. trigram ($n=3$)：条件2文字、$P(c \mid a, b)$。
1. $n$-gram：条件 $n-1$ 文字。
